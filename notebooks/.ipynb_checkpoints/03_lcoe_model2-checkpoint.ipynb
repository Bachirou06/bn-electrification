# NB03 — LCOE Model & Least-Cost Technology Selection
**Benin Least-Cost Electrification Analysis**

## Purpose
This is the **core modelling notebook**. It takes the demand-enriched settlement file from NB02 and
computes the Levelised Cost of Electricity (LCOE) for every feasible technology at every settlement,
then selects the least-cost technology for each.

## What this notebook produces
- `lcoe_grid` — grid extension LCOE for all unelectrified settlements (USD/kWh)
- `lcoe_mg_solar`, `lcoe_mg_hybrid`, `lcoe_mg_hydro` — mini-grid LCOEs
- `lcoe_shs` — solar home system LCOE
- `least_cost_tech` — winning technology (argmin across all LCOEs)
- `GridRolloutPhase` — Phase 1/2/3 for grid settlements
- Final output files for the Streamlit dashboard

## Run order dependency
**Must run NB02 first.** Requires `data/processed/settlements_demand_*.geojson`.

## Key design decisions made in this notebook
1. **Option B routing** (Cell 7) — uses `cKDTree` to nearest electrified settlement instead of raw grid distance
2. **Demand threshold** (via `mini_grid.py`) — 15,000 kWh/yr minimum for solar MG (ESMAP-aligned)
3. **LFP battery** — 10-year replacement cycle instead of lead-acid 5-year
4. **HydroSHEDS fix** — head ≥ 5 m threshold applied inside `mini_grid.py`

## 3b. Grid Routing — Option B: Nearest Electrified Settlement

**What this cell does:**
Computes `dist_nearest_electrified_km` — the distance from each *unelectrified* settlement to the
nearest *already-electrified* settlement, using a `scipy.cKDTree` spatial index for speed.

**Why Option B instead of raw grid distance (`GridDistKm`)?**
`GridDistKm` in the VIDA data measures distance to the nearest MV backbone line.
But in practice, grid extension projects connect to the nearest *existing customer point*
(transformer + LV network), not to the raw MV backbone.
Option B better reflects real project costing — existing connections have the LV infrastructure
already in place, reducing the marginal cost by 15–30%.

**Technical note:** `cKDTree` operates on Euclidean (flat-earth) coordinates.
At Benin's latitude (6–12°N) and typical inter-settlement distances (<100 km),
the flat-earth approximation introduces < 0.5% error — acceptable for planning purposes.

The computed distances are stored in `gdf['dist_nearest_electrified_km']` and passed to
`add_grid_lcoe()` in the next cell.


```python
from scipy.spatial import cKDTree

# Define masks here (also used in later cells)
unelec_mask = gdf['elec_status'] == 'unelectrified'
elec_mask   = gdf['elec_status'] == 'electrified'

# Use lon/lat columns directly — avoids CRS centroid warning
# These were added in NB00 from geometry centroids
if 'lon' not in gdf.columns:
    gdf['lon'] = gdf.geometry.centroid.x
    gdf['lat'] = gdf.geometry.centroid.y

elec_coords   = gdf.loc[elec_mask,   ['lon','lat']].values
unelec_coords = gdf.loc[unelec_mask, ['lon','lat']].values

# Build spatial index and find nearest electrified settlement
tree         = cKDTree(elec_coords)
dists_deg, _ = tree.query(unelec_coords, k=1)
dists_km     = dists_deg * 111.0  # degrees → km

gdf.loc[unelec_mask,  'dist_nearest_electrified_km'] = dists_km
gdf.loc[~unelec_mask, 'dist_nearest_electrified_km'] = 0.0

# ── Compare GridDistKm vs Option B ───────────────────────────────────────────
u = gdf[unelec_mask]
print('=== GRID DISTANCE COMPARISON ===')
print(f'\n  GridDistKm (to transmission line):')
print(f'    median = {u["GridDistKm"].median():.1f} km')
print(f'    mean   = {u["GridDistKm"].mean():.1f} km')
print(f'\n  dist_nearest_electrified_km (Option B):')
print(f'    median = {u["dist_nearest_electrified_km"].median():.1f} km')
print(f'    mean   = {u["dist_nearest_electrified_km"].mean():.1f} km')
print(f'    min    = {u["dist_nearest_electrified_km"].min():.1f} km')

closer = (u['dist_nearest_electrified_km'] < u['GridDistKm']).sum()
print(f'\n  Closer to electrified node than to grid: {closer:,} ({closer/len(u)*100:.1f}%)')

print(f'\n  Option B distance distribution:')
for km in [1, 2, 5, 10, 25, 50]:
    n = (u['dist_nearest_electrified_km'] <= km).sum()
    print(f'    <= {km:2d} km : {n:,} ({n/len(u)*100:.1f}%)')

# Comparison plot
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
fig.suptitle('Grid Distance: Transmission Line vs Nearest Electrified Settlement',
             fontsize=12, fontweight='bold')
axes[0].hist(u['GridDistKm'].clip(upper=150), bins=60,
             color='#E53935', alpha=0.7, label='GridDistKm (transmission line)')
axes[0].hist(u['dist_nearest_electrified_km'].clip(upper=150), bins=60,
             color='#1565C0', alpha=0.7, label='Option B (electrified node)')
axes[0].axvline(u['GridDistKm'].median(), color='#E53935', ls='--', lw=1.5)
axes[0].axvline(u['dist_nearest_electrified_km'].median(), color='#1565C0', ls='--', lw=1.5)
axes[0].set_xlabel('Distance (km)'); axes[0].set_ylabel('Settlements')
axes[0].set_title('Distance Distribution'); axes[0].legend(fontsize=8)

axes[1].scatter(u['GridDistKm'].clip(upper=150),
                u['dist_nearest_electrified_km'].clip(upper=150),
                alpha=0.2, s=3, color='#7B1FA2')
axes[1].plot([0,150],[0,150], 'r--', lw=1, label='Equal distance')
axes[1].set_xlabel('GridDistKm'); axes[1].set_ylabel('dist_nearest_electrified_km')
axes[1].set_title('Option B vs GridDistKm\n(below diagonal = closer via electrified node)')
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.savefig('../data/outputs/maps/grid_distance_comparison.png', dpi=150)
plt.show()
```

## 3. Grid Extension LCOE

Calls `add_grid_lcoe(gdf, dist_col='dist_nearest_electrified_km')` from `src/costs/grid_extension.py`.

**What is computed per settlement:**

| New column | Description |
|---|---|
| `grid_penalty` | OnSSET 5-factor terrain multiplier (≥ 1.0) |
| `mv_cost_per_km` | Slope-adjusted MV cost ($40k, $45k, or $55k/km) |
| `grid_capex_usd` | Total CAPEX = line_cost × penalty + $400/HH × connections |
| `lcoe_grid` | Grid LCOE (USD/kWh) — `np.inf` if infeasible |
| `grid_feasible` | True if distance ≤ `max_viable_distance_km` from config |

**Interpretation:** A settlement with `lcoe_grid = np.inf` will never win grid extension —
it is automatically excluded from the argmin in the least-cost selection step.
The LCOE comparison naturally handles the feasibility cutoff without any hard distance threshold.


```python
unelec_mask = gdf['elec_status'] == 'unelectrified'
print(f'Grid extension LCOE for {unelec_mask.sum():,} unelectrified settlements...')

# Use num_households_yT for grid sizing — grid is built to serve year-T demand
# (grid infrastructure has a 30yr lifetime; size for future load, not today's)
if 'num_households_yT' in gdf.columns:
    gdf_for_grid = gdf.copy()
    gdf_for_grid.loc[unelec_mask, 'num_households'] = gdf_for_grid.loc[unelec_mask, 'num_households_yT']
    gdf_for_grid = add_grid_lcoe(gdf_for_grid, dist_col='dist_nearest_electrified_km')
    # Copy back only the new LCOE columns
    for col in ['lcoe_grid','grid_capex_usd','grid_penalty','mv_cost_per_km','grid_feasible']:
        if col in gdf_for_grid.columns:
            gdf[col] = gdf_for_grid[col]
else:
    gdf = add_grid_lcoe(gdf)  # auto-detects dist_nearest_electrified_km

print(f'Grid LCOE computed ✓')
print(f'  Feasible: {(gdf["lcoe_grid"] < np.inf).sum():,}')
print(f'  Median LCOE (feasible): ${gdf["lcoe_grid"].replace(np.inf, np.nan).dropna().median():.3f}/kWh')
```

## 5. SHS LCOE

Calls `add_shs_lcoe(gdf)` from `src/costs/shs.py`.

SHS is the **universal fallback technology** — it is always feasible (no minimum demand, no distance
constraint). Every settlement will have a finite `lcoe_shs`.

**Kit assignment:**
- `mean_rwi ≥ 0.5` → Tier-3 kit ($350, 365 kWh/yr, 7-year lifetime)
- `mean_rwi < 0.5` → Tier-2 kit ($150, 73 kWh/yr, 5-year lifetime) — the vast majority

**Capacity cap:** The LCOE denominator is capped at the kit's rated output.
A Tier-2 kit at $150 cannot deliver more than 73 kWh/year — if VIDA demand per household
exceeds this, the kit is the binding constraint, not demand.
This prevents artificially low LCOE for high-demand settlements that would need multiple kits.

**Replacement costs** are included in the LCOE: Tier-2 kits are replaced at years 5 and 10
within the 15-year planning horizon.


```python
print(f'SHS LCOE for {unelec_mask.sum():,} unelectrified settlements...')
gdf_shs = gdf[unelec_mask].copy()
gdf_shs = add_shs_lcoe(gdf_shs)

# Merge back
for col in ['lcoe_shs','shs_capex_usd']:
    if col in gdf_shs.columns:
        if col not in gdf.columns: gdf[col] = np.inf
        gdf[col] = gdf[col].astype(float)
        gdf.loc[unelec_mask, col] = gdf_shs[col].astype(float).values
for col in ['shs_kit']:
    if col in gdf_shs.columns:
        if col not in gdf.columns: gdf[col] = ''
        gdf[col] = gdf[col].astype(str)
        gdf.loc[unelec_mask, col] = gdf_shs[col].astype(str).values
for col in ['shs_feasible']:
    if col in gdf_shs.columns:
        if col not in gdf.columns: gdf[col] = False
        gdf.loc[unelec_mask, col] = gdf_shs[col].values
if 'lcoe_shs' not in gdf.columns: gdf['lcoe_shs'] = np.inf
gdf['lcoe_shs'] = gdf['lcoe_shs'].astype(float)
gdf.loc[~unelec_mask, 'lcoe_shs'] = np.inf

# LCOE summary
print('\n=== LCOE SUMMARY (unelectrified) ===')
for col, lbl in [('lcoe_grid','Grid'),('lcoe_mg_solar','MG Solar'),
                  ('lcoe_mg_hybrid','MG Hybrid'),('lcoe_mg_hydro','MG Hydro'),
                  ('lcoe_shs','SHS')]:
    if col in gdf.columns:
        vals = gdf.loc[unelec_mask, col].replace(np.inf,np.nan).dropna()
        feas = (gdf.loc[unelec_mask, col] < np.inf).sum()
        med  = f'{vals.median():.3f}' if len(vals) > 0 else 'n/a'
        print(f'  {lbl:<12}: median=${med}/kWh  feasible={feas:,} ({feas/unelec_mask.sum()*100:.0f}%)')
```

```python
# ── SHS LCOE with price decline curve ────────────────────────────────────────
print(f'SHS LCOE for {unelec_mask.sum():,} unelectrified settlements...')
print(f'  Tier-2 price 2025/2040: ${gdf["shs_price_t2_y0"].iloc[0]:.0f} / ${gdf["shs_price_t2_yT"].iloc[0]:.0f}')

gdf_shs = gdf[unelec_mask].copy()

# Pass the declining price into shs.py by temporarily updating config
# The shs_lcoe_vida() function reads SHS_PARAMS from config, which was updated
# by the price_decline block — no override needed if config is set correctly.
# We verify the price decline is active:
from src.costs.shs import add_shs_lcoe
gdf_shs = add_shs_lcoe(gdf_shs)

# Merge back
for col in ['lcoe_shs','shs_kit','shs_capex_usd','shs_feasible']:
    if col in gdf_shs.columns:
        gdf[col] = gdf_shs[col].reindex(gdf.index)
gdf['lcoe_shs'] = gdf['lcoe_shs'].fillna(np.inf)

print(f'SHS LCOE computed ✓')
```

## 6. Least-Cost Selection

Calls `select_least_cost(gdf)` from `src/least_cost/technology_selector.py`.

**Algorithm:**
1. Build a matrix of shape `(n_settlements × 5 technologies)` — each cell is an LCOE value
2. Infeasible technologies = `np.inf` (auto-excluded)

**New columns added:**
- `least_cost_tech` — winning technology label
- `least_cost_lcoe` — LCOE of winning technology (USD/kWh)
- `least_cost_capex` — CAPEX of winning technology (USD)
- `n_feasible_techs` — number of technologies with finite LCOE (diagnostic)

## 6b. Three-Timestep Electrification with Grid Creep

Runs `run_timestep_electrification()` — a dynamic model that simulates grid expansion
across three 5-year periods aligned with Benin's national access targets.

### What happens at each timestep

| Step | Action |
|---|---|
| 1 | Update `dist_nearest_electrified_km` using ALL settlements connected so far |
| 2 | Re-run grid LCOE with updated (shorter) distances for remaining candidates |
| 3 | Score and rank candidates by priority (connections × demand / distance) |
| 4 | Select top-N candidates to meet this period's access target |
| 5 | Run argmin LCOE on winners — technology may differ from year-0 assignment |
| 6 | Mark winners as electrified → they become new grid nodes for the next period |

### Grid creep effect

A settlement 30 km from today's grid might be 8 km away in 2035 if a Phase 1
neighbour was electrified in 2030. Without the update, that settlement would
stay on SHS; with it, grid extension becomes viable in Phase 2.

### Access targets

| Phase | Year | National access rate | New settlements |
|---|---|---|---|
| Phase 1 | 2030 | 60 % | ~2,441 highest-priority |
| Phase 2 | 2035 | 85 % | ~4,301 next-priority |
| Phase 3 | 2040 | 100 % | ~2,581 remaining |

> **Note:** Technology assignments from Cell 17 (year-0) are overwritten here
> for winners of each phase. The final `least_cost_tech` column reflects the
> technology that won **at the time of connection**, accounting for grid creep.


```python
# ── Three-timestep electrification with grid creep ──────────────────────────
# Overwrites ElecTargetPhase, least_cost_tech, dist_at_connection_km
# for all unelectrified settlements.

gdf = run_timestep_electrification(gdf, verbose=True)

# ── Milestone summary ─────────────────────────────────────────────────────────
print()
print('=== ELECTRIFICATION MILESTONE SUMMARY ===')
ms = milestone_summary(gdf)
print(ms.to_string(index=False))

# ── Phase breakdown by technology ─────────────────────────────────────────────
print()
print('=== PHASE × TECHNOLOGY BREAKDOWN ===')
for phase in ['Phase 1 (→2030)', 'Phase 2 (→2035)', 'Phase 3 (→2040)']:
    sub = gdf[gdf['ElecTargetPhase'] == phase]
    if len(sub) == 0:
        continue
    print(f'  {phase} ({len(sub):,} settlements):')
    for tech, n in sub['least_cost_tech'].value_counts().items():
        pct = n / len(sub) * 100
        print(f'    {tech:<40} {n:5,} ({pct:.1f}%)')

# ── Grid creep diagnostic ─────────────────────────────────────────────────────
print()
print('=== GRID CREEP EFFECT ===')
grid_won = gdf[gdf['least_cost_tech'] == 'Grid Extension']
if 'dist_at_connection_km' in gdf.columns and len(grid_won) > 0:
    print(f'  Grid extension settlements  : {len(grid_won):,}')
    for phase in ['Phase 1 (→2030)', 'Phase 2 (→2035)', 'Phase 3 (→2040)']:
        sub = grid_won[grid_won['ElecTargetPhase'] == phase]
        if len(sub) > 0:
            med_dist = sub['dist_at_connection_km'].median()
            print(f'  {phase}: median dist at connection = {med_dist:.1f} km  '
                  f'({len(sub):,} settlements)')
```

## 6c. Population Growth — New Households by 2040

The dual-rate demand model (NB02) computed `hh_growth_total` for each
unelectrified settlement: the number of **additional households** that will
form by 2040 due to population growth (2.7%/yr × 40% rural share ≈ 1.1%/yr).

These new households are served by the same least-cost technology as the
settlement they form in — no separate technology selection needed.
Their CAPEX is at the *marginal connection cost* (no new MV line needed).


```python
# ── Population Growth CAPEX ────────────────────────────────────────────────────
# New HH in unelectrified settlements need connections but no new MV line
# Average marginal cost: SHS $150/HH, MG connection $280/HH, Grid $400/HH
# Weighted by technology distribution → ~$200/HH average

from src.config import GENERAL

T = GENERAL['planning_horizon_years']
unelec = gdf[gdf['elec_status'] == 'unelectrified'].copy()

if 'hh_growth_total' in unelec.columns:
    total_new_hh  = unelec['hh_growth_total'].sum()

    # Weight avg marginal cost by technology
    tech_weights = unelec.groupby('least_cost_tech')['hh_growth_total'].sum()
    marginal_costs = {
        'SHS': 150,
        'Mini-Grid: Solar PV Only': 280,
        'Mini-Grid: Mini-Hydro': 350,
        'Mini-Grid: Solar-Diesel Hybrid': 300,
        'Grid Extension': 400,
    }
    weighted_cost = sum(
        tech_weights.get(tech, 0) * cost
        for tech, cost in marginal_costs.items()
    ) / max(total_new_hh, 1)

    pop_growth_capex = total_new_hh * weighted_cost

    print("=== POPULATION GROWTH CONNECTIONS ===")
    print(f"  New HH in unelec settlements by {GENERAL['base_year']+T}: {total_new_hh:,.0f}")
    print(f"  Weighted marginal cost/HH: ${weighted_cost:.0f}")
    print(f"  Population growth CAPEX:   ${pop_growth_capex/1e6:.1f}M")
    print()
    print("  By technology:")
    for tech, hh in tech_weights.items():
        print(f"    {tech:<40}: {hh:,.0f} new HH")
else:
    print("hh_growth_total not found — run NB02 with dual-rate model first")
    total_new_hh = 0
    pop_growth_capex = 0
    weighted_cost = 200
```

## 9b. Full Programme Cost — Generation, T&D, OPEX, Institutions

This cell assembles the **complete programme cost** by adding the components
that the settlement-level LCOE model intentionally excludes:

| Component | Included in LCOE model? | Module |
|---|---|---|
| SHS / MG / Grid access CAPEX | ✅ Yes | NB03 cells 3–6 |
| Grid densification CAPEX | ✅ Yes (cell 3c) | grid_extension.py |
| Population growth connections | ✅ Yes (cell 6c) | demand_estimator.py |
| Generation build-out | ❌ No | generation_investment.py |
| T&D backbone reinforcement | ❌ No | generation_investment.py |
| OPEX NPV over lifetime | ❌ No (LCOE only) | generation_investment.py |
| Institutional / programme costs | ❌ No | generation_investment.py |


```python
# ── Full Programme Cost ─────────────────────────────────────────────────────────
import sys, os
sys.path.insert(0, str(Path(os.getcwd()).parent))

from src.costs.generation_investment import compute_programme_cost, print_programme_summary

# Gather inputs from this notebook
access_capex = gdf['least_cost_capex'].sum()
densif_capex = gdf.get('densification_capex', 0)
if hasattr(densif_capex, 'sum'):
    densif_capex = densif_capex.sum()

total_access_capex = access_capex + densif_capex

# New annual demand at year 15 from all newly connected settlements
unelec_mask = gdf['elec_status'] == 'unelectrified'
new_demand_kwh = gdf.loc[unelec_mask, 'demand_yearT_kwh'].sum()

# New HH from population growth
hh_growth = gdf.loc[unelec_mask, 'hh_growth_total'].sum() if 'hh_growth_total' in gdf.columns else 0

result = compute_programme_cost(
    access_capex_usd        = total_access_capex,
    new_annual_demand_kwh   = new_demand_kwh,
    hh_growth_total         = hh_growth,
    avg_capex_per_new_hh    = weighted_cost if 'weighted_cost' in dir() else 200,
    institutional_budget_usd= 80_000_000,
    productive_use_budget_usd= 60_000_000,
)

print_programme_summary(result)

# Store for dashboard
gdf.attrs['programme_cost'] = result
```

```python
from datetime import datetime
from pathlib import Path
from src.least_cost.technology_selector import technology_summary, milestone_summary

ts        = datetime.now().strftime('%Y%m%d_%H%M')
OUT_DIR   = Path('..') / 'data' / 'processed'
TABLE_DIR = Path('..') / 'data' / 'outputs' / 'tables'
TABLE_DIR.mkdir(parents=True, exist_ok=True)

# ── 1. Save full GeoJSON ──────────────────────────────────────────────────────
save_cols = [c for c in gdf.columns if c != 'demand_timeseries']
gdf[save_cols].to_file(OUT_DIR / f'settlements_lcoe_{ts}.geojson', driver='GeoJSON')
print(f'GeoJSON saved : settlements_lcoe_{ts}.geojson')

# ── 2. Technology summary CSV ─────────────────────────────────────────────────
summary = technology_summary(gdf)
summary.to_csv(TABLE_DIR / f'technology_summary_{ts}.csv', index=False)
print(f'Summary CSV   : technology_summary_{ts}.csv')
print(summary.to_string(index=False))

# ── 3. Milestone summary CSV (2030 / 2035 / 2040 targets) ────────────────────
ms = milestone_summary(gdf)
ms.to_csv(TABLE_DIR / f'milestone_summary_{ts}.csv', index=False)
print(f'\nMilestone CSV : milestone_summary_{ts}.csv')
print(ms.to_string(index=False))

# ── 4. Streamlit-ready flat CSV ───────────────────────────────────────────────
# One row per settlement, key fields renamed to OnSSET conventions
streamlit_cols = {
    # Identity
    'geometry'                    : 'geometry',
    'lon'                         : 'X_deg',
    'lat'                         : 'Y_deg',
    # Settlement characteristics
    'num_connections'             : 'NumConnections',
    'num_buildings'               : 'NumBuildings',
    'population'                  : 'Population',
    'mean_rwi'                    : 'WealthIndex',
    'elec_status'                 : 'ElecStatus',
    # Demand
    'demand_year0_kwh'            : 'DemandKWh_Y0',
    # Distances
    'GridDistKm'                  : 'GridDistKm',
    'dist_nearest_electrified_km' : 'DistNearElecKm',
    'dist_road_km'                : 'RoadDistKm',
    'dist_river_km'               : 'RiverDistKm',
    'DistSubstation'              : 'SubstationDistKm',
    # Terrain
    'Elevation'                   : 'ElevationM',
    'Slope'                       : 'SlopeDeg',
    'GHI'                         : 'GHI',
    'DieselPrice'                 : 'DieselPriceUSD_L',
    # Hydro
    'HydroHead'                   : 'HydroHeadM',
    'HydroDischarge'              : 'HydroDischargeM3s',
    # Grid
    'grid_penalty'                : 'GridPenalty',
    'mv_cost_per_km'              : 'MVCostPerKm',
    # LCOE by technology
    'lcoe_grid'                   : 'LCOE_Grid',
    'lcoe_mg_solar'               : 'LCOE_MG_Solar',
    'lcoe_mg_hybrid'              : 'LCOE_MG_Hybrid',
    'lcoe_mg_hydro'               : 'LCOE_MG_Hydro',
    'lcoe_shs'                    : 'LCOE_SHS',
    # Technology selection
    'least_cost_tech'             : 'MinimumOverall',
    'least_cost_lcoe'             : 'MinimumOverallLCOE',
    'least_cost_capex'            : 'InvestmentCost',
    # CAPEX breakdown
    'grid_capex_usd'              : 'CAPEX_Grid',
    'minigrid_capex_usd'          : 'CAPEX_MG',
    'minigrid_peak_kw'            : 'CapacityKW',
    'shs_capex_usd'               : 'CAPEX_SHS',
    # ── ELECTRIFICATION TARGET PHASES (2030 / 2035 / 2040) ───────────────────
    'ElecTargetPhase'             : 'ElecTargetPhase',   # 'Phase 1 (→2030)' etc.
    'ElecTargetYear'              : 'ElecTargetYear',    # 2030 / 2035 / 2040
    'rollout_rank'                : 'RolloutRank',       # 1 = highest priority
    'rollout_priority'            : 'RolloutPriority',   # raw score
    # Grid-only sub-phases
    'GridRolloutPhase'            : 'GridRolloutPhase',  # within grid settlements
    # Grid creep — distance at time of connection (updated each timestep)
    'dist_at_connection_km'       : 'DistAtConnectionKm',
}

# Only include columns that actually exist in gdf
valid_cols = {k: v for k, v in streamlit_cols.items() if k in gdf.columns or k == 'geometry'}
df_out = gdf[[c for c in valid_cols.keys() if c in gdf.columns]].rename(columns=valid_cols)

# Save CSV (drop geometry for flat format)
df_csv = df_out.drop(columns=['geometry'], errors='ignore')
csv_path = TABLE_DIR / f'benin_electrification_streamlit_{ts}.csv'
df_csv.to_csv(csv_path, index=False)
print(f'\nStreamlit CSV : benin_electrification_streamlit_{ts}.csv')
print(f'  Rows: {len(df_csv):,}  |  Columns: {len(df_csv.columns)}')
if 'ElecTargetPhase' in df_csv.columns:
    print('  ElecTargetPhase distribution:')
    print(df_csv['ElecTargetPhase'].value_counts().to_string())
else:
    print('  WARNING: ElecTargetPhase not in output — check add_rollout_phases() was called')

# Save GeoJSON (with geometry)
if 'geometry' in df_out.columns:
    import geopandas as gpd
    gdf_out = gpd.GeoDataFrame(df_out, geometry='geometry', crs=gdf.crs)
    geojson_path = TABLE_DIR / f'benin_electrification_streamlit_{ts}.geojson'
    gdf_out.to_file(geojson_path, driver='GeoJSON')
    print(f'Streamlit GeoJSON: benin_electrification_streamlit_{ts}.geojson')

print(f'\n✓ All outputs saved to {TABLE_DIR}')
```
