# 03 — LCOE Model & Least-Cost Technology Selection
**Benin Least-Cost Electrification Analysis**

Loads the demand + electrification status output from Notebook 02 and computes
the least-cost technology for each settlement.

**Routing logic:**
| `elec_status` | `connection_rate` | Action |
|---|---|---|
| `electrified` | >= 90% | Already at target — no intervention |
| `electrified` | < 90% | **Grid densification** — LV extension within settlement |
| `unelectrified` | N/A | **Least-cost selection** — grid extension / MG / SHS |

**Sections:**
1. Load data & routing
2. Cost assumptions
3. Grid densification (electrified settlements)
4. Grid extension LCOE (unelectrified)
5. Mini-grid LCOE — solar, hybrid, hydro
6. SHS LCOE
7. Least-cost selection
8. Results map
9. Save


```python
# Grid densification check
# In VIDA, num_connections = satellite-detected rooftops (IS the building count)
# so connection_rate = num_connections/num_buildings = 1.0 for all settlements
# Densification via this model is not applicable — VIDA has no separate
# 'unconnected buildings' signal. Electrified settlements are already served.

gdf['connection_rate']      = 1.0
gdf['HH_to_connect']         = 0
gdf['densification_needed']  = False
gdf['densification_capex']   = 0.0
gdf['lcoe_densification']    = np.nan

elec = gdf[gdf['elec_status'] == 'electrified']
print(f'Electrified settlements : {len(elec):,} — already served, no densification needed')
print(f'  (VIDA num_connections = satellite rooftops = building count)')
print(f'  (connection_rate = 1.0 by definition in VIDA data)')
```

## 1. Load Data & Routing


```python
PROCESSED_DIR = Path('..') / 'data' / 'processed'

# Auto-detect newest timestamped demand file from NB02
candidates = sorted(PROCESSED_DIR.glob('settlements_demand_*.geojson'), reverse=True)
if not candidates:
    raise FileNotFoundError('No settlements_demand_*.geojson found — run NB02 first')

DEMAND_PATH = candidates[0]
print(f'Loading: {DEMAND_PATH.name}')
gdf = gpd.read_file(DEMAND_PATH)
print(f'Loaded {len(gdf):,} settlements  |  {len(gdf.columns)} columns')

# Deserialise demand timeseries from string (GeoJSON serialises lists as strings)
if 'demand_timeseries' in gdf.columns:
    if isinstance(gdf['demand_timeseries'].iloc[0], str):
        gdf['demand_timeseries'] = gdf['demand_timeseries'].apply(ast.literal_eval)
    print('demand_timeseries deserialised ✓')
else:
    # Reconstruct from year0 and yearT using linear interpolation
    T = GENERAL['planning_horizon_years']
    g = 0.04
    gdf['demand_timeseries'] = gdf['demand_year0_kwh'].apply(
        lambda d0: [d0 * (1+g)**t for t in range(T+1)]
    )
    print('demand_timeseries reconstructed from year0 ✓')

# ── Routing summary ───────────────────────────────────────────────────────────
print(f'\nElectrification status:')
for status, n in gdf['elec_status'].value_counts().items():
    print(f'  {status:<15}: {n:,} ({n/len(gdf)*100:.1f}%)')

# Column checks
key_cols = ['elec_status','demand_year0_kwh','demand_timeseries',
            'GridDistKm','Slope','Elevation','LandCover',
            'dist_road_km','DistSubstation','GHI','DieselPrice',
            'dist_river_km','HydroHead','HydroDischarge','num_connections']
print('\nKey column check:')
for col in key_cols:
    status = '✓' if col in gdf.columns else '✗ MISSING'
    print(f'  {col:<25}: {status}')
```

## 2. Cost Assumptions


```python
from src.config import CONFIG, GRID, DEMAND as DEMAND_CFG

# Densification parameters
TARGET_CONNECTION_RATE = 0.90   # ABERME Plan Directeur 2022 target
LV_COST_PER_KM        = 8_000   # USD/km — ABERME LV 400V unit cost
LV_LENGTH_PER_HH      = 0.030   # km — 30m LV per new connection (OnSSET cluster model)
CONNECTION_COST       = 400     # USD/HH — ABERME rural programme budget
DENSIF_COST_PER_HH    = LV_COST_PER_KM * LV_LENGTH_PER_HH + CONNECTION_COST  # = $640

print('=== COST ASSUMPTIONS ===')
print(f'\nGrid Extension:')
print(f'  MV cost/km (slope-adjusted): $40,000-$55,000')
print(f'  Connection cost             : ${GRID["capex_per_connection"]:,}/HH')
print(f'  Grid penalty                : OnSSET terrain model (slope x0.30, substation x0.20, ...)')
print(f'  Max viable distance         : {GRID["max_viable_distance_km"]} km')
print(f'  Loss rate                   : {GRID["loss_rate"]*100:.0f}%')
print(f'  Lifetime                    : {GRID["lifetime_years"]} years')
print(f'\nGrid Densification:')
print(f'  Target connection rate      : {TARGET_CONNECTION_RATE*100:.0f}%  (ABERME 2022)')
print(f'  LV cost/km                  : ${LV_COST_PER_KM:,}/km')
print(f'  LV length per HH            : {LV_LENGTH_PER_HH*1000:.0f}m')
print(f'  Connection cost             : ${CONNECTION_COST:,}/HH')
print(f'  Total cost per HH           : ${DENSIF_COST_PER_HH:,}/HH')
print(f'\nGeneral:')
print(f'  Discount rate               : {DISCOUNT_RATE*100:.0f}%')
print(f'  Planning horizon            : {GENERAL["planning_horizon_years"]} years')
```

## 3. Grid Densification (Electrified Settlements)

For settlements already connected to the grid but with connection rate < 90%,
compute the cost of connecting remaining households via LV network extension
within the settlement boundary.

**Formula:** `CAPEX = $640 x HH_to_connect`  
**Source:** ABERME (2022) LV unit costs + OnSSET cluster model


```python
# Grid densification — electrified settlements with connection gap
gdf = add_densification(gdf)
```

## 4. Grid Extension LCOE (Unelectrified Settlements)

Grid extension LCOE with:
- **Slope-adjusted MV cost**: $40k/km (flat) → $45k/km (moderate) → $55k/km (steep)
- **OnSSET terrain penalty**: 5-factor score (slope x0.30, substation x0.20, land cover x0.20, road dist x0.15, elevation x0.15)
- **Grid feasibility**: settlements > 25km from grid → LCOE = inf


```python
# Run only on unelectrified settlements — electrified ones already have grid
unelec_mask = gdf['elec_status'] == 'unelectrified'
print(f'Computing grid extension LCOE for {unelec_mask.sum():,} unelectrified settlements...')

gdf_unelec = gdf[unelec_mask].copy()
gdf_unelec = add_grid_lcoe(gdf_unelec, dist_col='GridDistKm')

# Merge back into full dataset
for col in ['grid_penalty','mv_cost_per_km','grid_capex_usd','lcoe_grid','grid_feasible']:
    if col in gdf_unelec.columns:
        gdf.loc[unelec_mask, col] = gdf_unelec[col].values

# Electrified settlements: grid cost = 0 (already connected)
gdf.loc[~unelec_mask, 'lcoe_grid']     = 0.0
gdf.loc[~unelec_mask, 'grid_feasible'] = True
gdf.loc[~unelec_mask, 'grid_capex_usd'] = 0.0

print(f'\n=== GRID EXTENSION RESULTS ===')
grid_fin = gdf.loc[unelec_mask, 'lcoe_grid'].replace(np.inf, np.nan).dropna()
print(f'  Feasible settlements  : {gdf.loc[unelec_mask,"grid_feasible"].sum():,} / {unelec_mask.sum():,}')
print(f'  Median LCOE           : ${grid_fin.median():.3f}/kWh')
print(f'  Mean LCOE             : ${grid_fin.mean():.3f}/kWh')
print(f'  Total CAPEX (feasible): ${gdf.loc[unelec_mask & gdf["grid_feasible"],"grid_capex_usd"].sum()/1e6:.1f}M')
print(f'\n  Grid penalty stats (unelectrified):')
if 'grid_penalty' in gdf.columns:
    pen = gdf.loc[unelec_mask,'grid_penalty'].dropna()
    print(f'    min={pen.min():.4f}  mean={pen.mean():.4f}  max={pen.max():.4f}')
print(f'\n  MV cost/km distribution:')
if 'mv_cost_per_km' in gdf.columns:
    print(gdf.loc[unelec_mask,'mv_cost_per_km'].value_counts().sort_index().to_string())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Grid Extension — Terrain Penalty & LCOE', fontsize=13, fontweight='bold')

unelec = gdf[gdf['elec_status'] == 'unelectrified']

# 1. Grid penalty distribution
if 'grid_penalty' in unelec.columns:
    axes[0].hist(unelec['grid_penalty'].dropna(), bins=50,
                 color='#1565C0', edgecolor='white')
    axes[0].set_xlabel('Grid penalty multiplier')
    axes[0].set_ylabel('Settlements')
    axes[0].set_title('OnSSET Terrain Penalty\n(1.0 = no penalty)')
    med = unelec['grid_penalty'].median()
    axes[0].axvline(med, color='red', ls='--', label=f'Median={med:.4f}')
    axes[0].legend(fontsize=8)

# 2. MV cost/km by slope tier
if 'mv_cost_per_km' in unelec.columns:
    counts = unelec['mv_cost_per_km'].value_counts().sort_index()
    labels = [str(int(v/1000)) + 'k' for v in counts.index]
    bars = axes[1].bar(labels, counts.values,
                       color=['#4CAF50','#FF9800','#F44336'][:len(counts)])
    axes[1].set_title('MV Cost Tier\n(slope-adjusted, ABERME)')
    axes[1].set_ylabel('Settlements')
    for bar, (v, n) in zip(bars, counts.items()):
        axes[1].text(bar.get_x()+bar.get_width()/2, bar.get_height()+20,
                     f'{n:,}\n({n/len(unelec)*100:.1f}%)', ha='center', fontsize=9)

# 3. Grid LCOE vs distance
feasible = unelec[unelec.get('grid_feasible', pd.Series(True, index=unelec.index)) == True]
if 'lcoe_grid' in feasible.columns and 'GridDistKm' in feasible.columns:
    lcoe_vals = feasible['lcoe_grid'].replace(np.inf, np.nan).dropna()
    axes[2].scatter(feasible.loc[lcoe_vals.index, 'GridDistKm'],
                    lcoe_vals, alpha=0.2, s=3, color='#1565C0')
    axes[2].set_xlabel('Grid distance (km)')
    axes[2].set_ylabel('Grid LCOE (USD/kWh)')
    axes[2].set_title('Grid LCOE vs Distance\n(feasible settlements only)')
    axes[2].set_ylim(0, lcoe_vals.quantile(0.95))

plt.tight_layout()
plt.savefig('../data/outputs/maps/grid_extension_analysis.png', dpi=150)
plt.show()
```

## 5. Save Grid & Densification Results
Saves intermediate results before computing MG and SHS in next steps.


```python
from datetime import datetime
ts = datetime.now().strftime('%Y%m%d_%H%M')
OUT_DIR   = Path('..') / 'data' / 'processed'
TABLE_DIR = Path('..') / 'data' / 'outputs' / 'tables'
TABLE_DIR.mkdir(parents=True, exist_ok=True)

# Summary table
summary = pd.DataFrame([
    {'Intervention': 'Grid densification',
     'Settlements'  : int(gdf['densification_needed'].sum()),
     'HH_affected'  : int(gdf['HH_to_connect'].sum()),
     'Total_CAPEX_M': round(gdf['densification_capex'].sum()/1e6, 1),
     'Median_LCOE'  : round(gdf['lcoe_densification'].dropna().median(), 3)},
    {'Intervention': 'Grid extension (feasible)',
     'Settlements'  : int(gdf['grid_feasible'].sum()),
     'HH_affected'  : int(gdf.loc[gdf['grid_feasible'],'num_connections'].sum()),
     'Total_CAPEX_M': round(gdf.loc[gdf['grid_feasible'],'grid_capex_usd'].sum()/1e6, 1),
     'Median_LCOE'  : round(gdf.loc[gdf['elec_status']=='unelectrified','lcoe_grid']
                            .replace(np.inf,np.nan).dropna().median(), 3)},
])
summary.to_csv(TABLE_DIR / f'grid_results_{ts}.csv', index=False)

print('=== GRID & DENSIFICATION SUMMARY ===')
print(summary.to_string(index=False))
print(f'\nSaved: grid_results_{ts}.csv')
print('\nReady for next step: Mini-grid LCOE (cells below)')
```

## 6. Mini-Grid LCOE

Three sub-types computed for all unelectrified settlements:
- **Solar PV only** — peak-to-base 3.5, min 20 connections
- **Solar-diesel hybrid** — peak-to-base 2.5, min 50 connections, road <= 50km, spatially varying `DieselPrice`
- **Mini-hydro** — peak-to-base 2.0, min 30 connections, river <= 5km, head >= 2m


```python
unelec_mask = gdf['elec_status'] == 'unelectrified'
gdf_unelec  = gdf[unelec_mask].copy()

print(f'Computing mini-grid LCOE for {unelec_mask.sum():,} unelectrified settlements...')
with warnings.catch_warnings(record=True) as w:
    warnings.simplefilter('always')
    gdf_unelec = add_minigrid_lcoe(gdf_unelec)
    for warning in w:
        print(f'  ⚠ {warning.message}')

# Merge back
mg_cols = ['lcoe_mg_solar','lcoe_mg_hybrid','lcoe_mg_hydro',
           'lcoe_minigrid','minigrid_subtype','minigrid_subtype_label',
           'minigrid_capex_usd','minigrid_peak_kw','minigrid_feasible','mg_hydro_feasible']
for col in mg_cols:
    if col in gdf_unelec.columns:
        gdf.loc[unelec_mask, col] = gdf_unelec[col].values

# Electrified: no mini-grid needed
for col in ['lcoe_mg_solar','lcoe_mg_hybrid','lcoe_mg_hydro','lcoe_minigrid']:
    gdf.loc[~unelec_mask, col] = np.inf

print(f'\nMini-hydro feasible: {gdf["mg_hydro_feasible"].sum():,} settlements')
print(f'Any MG feasible    : {gdf["minigrid_feasible"].sum():,} settlements')
```

## 7. SHS LCOE

SHS is always feasible as a last-resort option.
Kit assigned from `mean_rwi` after technology selection (post-selection label only).
LCOE denominator = VIDA `demand_timeseries / num_households`.


```python
print(f'Computing SHS LCOE for {unelec_mask.sum():,} unelectrified settlements...')
gdf_unelec_shs = gdf[unelec_mask].copy()
gdf_unelec_shs = add_shs_lcoe(gdf_unelec_shs)

for col in ['lcoe_shs','shs_capex_usd','shs_kit','shs_feasible']:
    if col in gdf_unelec_shs.columns:
        gdf.loc[unelec_mask, col] = gdf_unelec_shs[col].values

gdf.loc[~unelec_mask, 'lcoe_shs'] = np.inf

# ── LCOE summary across all technologies ─────────────────────────────────────
print(f'\n=== LCOE SUMMARY (unelectrified settlements) ===')
for col, lbl in [
    ('lcoe_grid',      'Grid Extension  '),
    ('lcoe_mg_solar',  'MG Solar        '),
    ('lcoe_mg_hybrid', 'MG Hybrid       '),
    ('lcoe_mg_hydro',  'MG Hydro        '),
    ('lcoe_shs',       'SHS             '),
]:
    if col in gdf.columns:
        vals = gdf.loc[unelec_mask, col].replace(np.inf, np.nan).dropna()
        feas = (gdf.loc[unelec_mask, col] < np.inf).sum()
        if len(vals) > 0:
            print(f'  {lbl}: median=${vals.median():.3f}  '
                  f'mean=${vals.mean():.3f}  '
                  f'feasible={feas:,} ({feas/unelec_mask.sum()*100:.0f}%)')
```

## 8. Least-Cost Technology Selection

For each unelectrified settlement, select the technology with the lowest LCOE
among all feasible options. Electrified settlements assigned 'Already Electrified'.


```python
gdf = select_least_cost(gdf)
gdf = add_grid_priority(gdf)

summary = technology_summary(gdf)
print('=== LEAST-COST TECHNOLOGY RESULTS ===')
print(summary.to_string(index=False))
print(f'\nTotal investment (unelectrified): '
      f'${gdf.loc[unelec_mask,"least_cost_capex"].sum()/1e9:.2f}B')

# SHS kit breakdown (post-selection)
shs_mask = gdf['least_cost_tech'] == 'SHS'
if shs_mask.sum() > 0:
    print(f'\nSHS kit assignment (from mean_rwi):')
    print(gdf.loc[shs_mask,'shs_kit'].value_counts().to_string())
```

## 9. Results Visualisation

```python
import matplotlib.patches as mpatches
from pathlib import Path

TECH_COLORS = {
    'Already Electrified'            : '#90CAF9',
    'Grid Extension'                 : '#1565C0',
    'Mini-Grid: Solar PV Only'       : '#FDD835',
    'Mini-Grid: Solar-Diesel Hybrid' : '#E53935',
    'Mini-Grid: Mini-Hydro'          : '#0288D1',
    'SHS'                            : '#43A047',
}
TECH_SIZES = {
    'Already Electrified': 3,
    'Grid Extension'     : 6,
    'Mini-Grid: Solar PV Only': 4,
    'Mini-Grid: Solar-Diesel Hybrid': 4,
    'Mini-Grid: Mini-Hydro': 5,
    'SHS'                : 2,
}

fig, ax = plt.subplots(figsize=(9, 11))
ax.set_facecolor('#E8F4F8')

# Country boundary
for bpath in [
    Path('..') / 'data' / 'raw' / 'Benin_boundary.gpkg',
    Path('..') / 'data' / 'raw' / 'benin_boundary.gpkg',
]:
    if bpath.exists():
        import geopandas as gpd
        gpd.read_file(bpath).plot(ax=ax, color='white', edgecolor='#333333',
                                   linewidth=1.2, zorder=1)
        break

# Transmission lines
TRANS = Path('..') / 'data' / 'raw' / 'Benin_existing_transmission_lines_2017.geojson'
if TRANS.exists():
    import geopandas as gpd
    gpd.read_file(TRANS).plot(ax=ax, color='#212121', linewidth=0.8, alpha=0.6, zorder=2)

# Plot order: electrified last so it sits on top
plot_order = ['SHS','Mini-Grid: Solar PV Only','Mini-Grid: Solar-Diesel Hybrid',
              'Mini-Grid: Mini-Hydro','Grid Extension','Already Electrified']
for tech in plot_order:
    sub = gdf[gdf['least_cost_tech'] == tech]
    if len(sub) == 0: continue
    sub.plot(ax=ax, color=TECH_COLORS.get(tech,'#999999'),
             markersize=TECH_SIZES.get(tech,3), alpha=0.7,
             marker='o', linewidth=0, zorder=3)

ax.set_xlim(0.8, 3.9)
ax.set_ylim(6.1, 12.5)

# Legend
counts = gdf['least_cost_tech'].value_counts()
legend_elements = [
    mpatches.Patch(color=TECH_COLORS.get(t,'#999'),
                   label=f'{t} ({counts.get(t,0):,} — {counts.get(t,0)/len(gdf)*100:.1f}%)')
    for t in plot_order if counts.get(t, 0) > 0
]
legend_elements.append(
    mpatches.Patch(color='#212121', label='Transmission lines')
)
ax.legend(handles=legend_elements, loc='lower left', fontsize=7,
          framealpha=0.95, edgecolor='gray')
ax.set_title(
    f'Benin — Least-Cost Electrification Technology\n'
    f'17,205 settlements | {len(gdf[gdf["elec_status"]=="unelectrified"]):,} unelectrified analysed',
    fontsize=11, fontweight='bold'
)
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
plt.tight_layout()
plt.savefig('../data/outputs/maps/technology_map.png', dpi=150, bbox_inches='tight')
plt.show()
print('Map saved ✓')
```

## 11. Save Results


```python
from datetime import datetime
from pathlib import Path
ts = datetime.now().strftime('%Y%m%d_%H%M')

OUT_DIR   = Path('..') / 'data' / 'processed'
TABLE_DIR = Path('..') / 'data' / 'outputs' / 'tables'
TABLE_DIR.mkdir(parents=True, exist_ok=True)

# Save full results GeoJSON
save_cols = [c for c in gdf.columns if c != 'demand_timeseries']
gdf[save_cols].to_file(OUT_DIR / f'settlements_lcoe_{ts}.geojson', driver='GeoJSON')
print(f'GeoJSON saved : settlements_lcoe_{ts}.geojson')

# Technology summary CSV
summary = technology_summary(gdf)
summary.to_csv(TABLE_DIR / f'technology_summary_{ts}.csv', index=False)
print(f'Summary saved : technology_summary_{ts}.csv')

# Grid priority CSV
grid_subs = gdf[gdf['least_cost_tech'] == 'Grid Extension'].copy()
if len(grid_subs) > 0:
    grid_cols = [c for c in ['village_name','GridDistKm','demand_year0_kwh',
                              'num_connections','grid_priority_score_norm',
                              'grid_rollout_phase','lcoe_grid','grid_capex_usd']
                 if c in grid_subs.columns]
    grid_subs[grid_cols].sort_values('grid_priority_score_norm', ascending=False
        ).to_csv(TABLE_DIR / f'grid_priority_{ts}.csv', index=False)
    print(f'Grid priority : grid_priority_{ts}.csv  ({len(grid_subs):,} settlements)')

print(f'\n=== FINAL SUMMARY ===')
print(summary.to_string(index=False))
```
