# 06 — Settlement Data Investigation
**Benin Least-Cost Electrification Analysis**

Deep dive into the settlement dataset covering:
- Population and size distribution
- Distance to grid and electrification potential
- Wealth (RWI) and demand tier spatial patterns
- Building density and structure types
- Social services distribution
- Correlations between key variables

```python
import sys
sys.path.append('..')

import geopandas as gpd
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from pathlib import Path

BASE_DIR = Path('..').resolve()
sns.set_theme(style='whitegrid', palette='muted')
pd.set_option('display.float_format', '{:.2f}'.format)

gdf = gpd.read_file(str(BASE_DIR / 'data' / 'processed' / 'settlements_demand.geojson'))
print(f'Loaded {len(gdf):,} settlements')
print(f'Departments: {sorted(gdf["admin_cgaz_1"].dropna().unique())}')
```

## 1. Population and Settlement Size Distribution

```python
bins   = [0, 100, 500, 2000, 10000, np.inf]
labels = ['< 100', '100-500', '500-2000', '2000-10000', '> 10000']
gdf['size_category'] = pd.cut(gdf['population'], bins=bins, labels=labels)

size_summary = gdf.groupby('size_category', observed=True).agg(
    N_settlements=('identifier', 'count'),
    Total_pop=('population', 'sum'),
    Avg_pop=('population', 'mean'),
).assign(
    Pct_settlements=lambda x: x['N_settlements'] / x['N_settlements'].sum() * 100,
    Pct_population=lambda x: x['Total_pop'] / x['Total_pop'].sum() * 100,
)

print('=== SETTLEMENT SIZE DISTRIBUTION ===')
print(size_summary.round(1).to_string())

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Population and Settlement Size Distribution', fontsize=13, fontweight='bold')

size_summary['N_settlements'].plot(kind='bar', ax=axes[0], color='#2196F3', edgecolor='white')
axes[0].set_title('Number of Settlements by Size')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=30)

size_summary['Pct_population'].plot(kind='bar', ax=axes[1], color='#FF9800', edgecolor='white')
axes[1].set_title('% of Total Population by Size')
axes[1].set_ylabel('% of Population')
axes[1].tick_params(axis='x', rotation=30)

axes[2].hist(np.log10(gdf['population'].clip(lower=1)), bins=50, color='#4CAF50', edgecolor='white')
axes[2].set_title('Population Distribution (log10 scale)')
axes[2].set_xlabel('log10(Population)')
axes[2].set_ylabel('Number of Settlements')

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data/outputs/maps/inv_population.png'), dpi=150)
plt.show()
```

## 2. Population by Department

```python
dept = gdf.groupby('admin_cgaz_1').agg(
    N_settlements=('identifier', 'count'),
    Total_pop=('population', 'sum'),
    Avg_pop=('population', 'mean'),
    Median_pop=('population', 'median'),
    Pct_nightlight=('has_nightlight', 'mean'),
).sort_values('Total_pop', ascending=False)
dept['Pct_nightlight'] = dept['Pct_nightlight'] * 100

print('=== SETTLEMENTS BY DEPARTMENT ===')
print(dept.round(1).to_string())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Settlement Distribution by Department', fontsize=13, fontweight='bold')

dept['Total_pop'].plot(kind='barh', ax=axes[0], color='#2196F3', edgecolor='white')
axes[0].set_title('Total Population by Department')
axes[0].set_xlabel('Population')

dept['N_settlements'].plot(kind='barh', ax=axes[1], color='#FF9800', edgecolor='white')
axes[1].set_title('Number of Settlements by Department')
axes[1].set_xlabel('Count')

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data/outputs/maps/inv_departments.png'), dpi=150)
plt.show()
```

## 3. Distance to Grid and Electrification Potential

```python
dist_col = 'distance_to_existing_transmission_lines'

dist_bins   = [0, 5, 10, 15, 25, 50, 100, np.inf]
dist_labels = ['0-5', '5-10', '10-15', '15-25', '25-50', '50-100', '>100']
gdf['dist_band'] = pd.cut(gdf[dist_col], bins=dist_bins, labels=dist_labels)

dist_summary = gdf.groupby('dist_band', observed=True).agg(
    N_settlements=('identifier', 'count'),
    Total_pop=('population', 'sum'),
    Pct_nightlight=('has_nightlight', 'mean'),
).assign(
    Pct_settlements=lambda x: x['N_settlements'] / x['N_settlements'].sum() * 100,
    Cumulative_pct=lambda x: (x['N_settlements'] / x['N_settlements'].sum() * 100).cumsum(),
)
dist_summary['Pct_nightlight'] *= 100

print('=== DISTANCE TO EXISTING GRID ===')
print(dist_summary.round(1).to_string())

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Distance to Existing Transmission Lines', fontsize=13, fontweight='bold')

dist_summary['N_settlements'].plot(kind='bar', ax=axes[0], color='#2196F3', edgecolor='white')
axes[0].axvline(3.5, color='red', linestyle='--', linewidth=1.5, label='25km threshold')
axes[0].set_title('Settlements by Distance Band')
axes[0].set_ylabel('Count')
axes[0].set_xlabel('Distance (km)')
axes[0].tick_params(axis='x', rotation=30)
axes[0].legend()

dist_summary['Total_pop'].plot(kind='bar', ax=axes[1], color='#FF9800', edgecolor='white')
axes[1].axvline(3.5, color='red', linestyle='--', linewidth=1.5)
axes[1].set_title('Population by Distance Band')
axes[1].set_ylabel('Population')
axes[1].set_xlabel('Distance (km)')
axes[1].tick_params(axis='x', rotation=30)

sorted_dist = gdf[dist_col].sort_values()
cdf = np.arange(1, len(sorted_dist) + 1) / len(sorted_dist) * 100
axes[2].plot(sorted_dist.values, cdf, color='#4CAF50', linewidth=2)
axes[2].axvline(25, color='red', linestyle='--', label='25km threshold')
axes[2].axvline(10, color='orange', linestyle='--', label='10km threshold')
axes[2].set_title('CDF of Distance to Grid')
axes[2].set_xlabel('Distance (km)')
axes[2].set_ylabel('% of Settlements')
axes[2].set_xlim(0, 150)
axes[2].legend()

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data/outputs/maps/inv_grid_distance.png'), dpi=150)
plt.show()

within_25 = (gdf[dist_col] <= 25).mean() * 100
within_10 = (gdf[dist_col] <= 10).mean() * 100
print(f'Settlements within 10km : {within_10:.1f}% - strong grid extension candidates')
print(f'Settlements within 25km : {within_25:.1f}% - potential grid extension')
print(f'Settlements beyond 25km : {100 - within_25:.1f}% - mini-grid or SHS candidates')
```

## 4. Wealth Index (RWI) Spatial Patterns

```python
print('=== RELATIVE WEALTH INDEX (RWI) ===')
print(gdf['mean_rwi'].describe().round(3))
print(f'Negative RWI (poor)    : {(gdf["mean_rwi"] < 0).mean()*100:.1f}%')
print(f'Positive RWI (wealthy) : {(gdf["mean_rwi"] > 0).mean()*100:.1f}%')

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Relative Wealth Index (RWI) Analysis', fontsize=13, fontweight='bold')

axes[0].hist(gdf['mean_rwi'].dropna(), bins=60, color='#9C27B0', edgecolor='white', alpha=0.8)
axes[0].axvline(0, color='red', linestyle='--', label='RWI = 0')
axes[0].axvline(gdf['mean_rwi'].median(), color='orange', linestyle='--',
                label=f'Median = {gdf["mean_rwi"].median():.2f}')
axes[0].set_title('RWI Distribution')
axes[0].set_xlabel('Mean RWI')
axes[0].legend()

rwi_dept = gdf.groupby('admin_cgaz_1')['mean_rwi'].median().sort_values()
colors = ['#E53935' if v < 0 else '#43A047' for v in rwi_dept.values]
rwi_dept.plot(kind='barh', ax=axes[1], color=colors, edgecolor='white')
axes[1].axvline(0, color='black', linewidth=1)
axes[1].set_title('Median RWI by Department')
axes[1].set_xlabel('Median RWI')

axes[2].scatter(
    gdf['distance_to_existing_transmission_lines'].clip(upper=100),
    gdf['mean_rwi'], alpha=0.1, s=5, color='#9C27B0'
)
axes[2].axhline(0, color='red', linestyle='--', alpha=0.5)
axes[2].axvline(25, color='orange', linestyle='--', alpha=0.5, label='25km threshold')
corr = gdf[['distance_to_existing_transmission_lines', 'mean_rwi']].corr().iloc[0, 1]
axes[2].set_title(f'RWI vs Distance to Grid (r={corr:.2f})')
axes[2].set_xlabel('Distance to Grid (km)')
axes[2].set_ylabel('Mean RWI')
axes[2].legend()

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data/outputs/maps/inv_rwi.png'), dpi=150)
plt.show()
```

## 5. Building Density and Structure Types

```python
print('=== BUILDING STRUCTURE ANALYSIS ===')
print(gdf[['num_buildings', 'small_buildings', 'medium_buildings',
           'large_buildings', 'very_small_structures',
           'Medium_and_large_buildings_pc', 'building_density_percent']].describe().round(2))

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Building Density and Structure Types', fontsize=13, fontweight='bold')

avg_mix = pd.Series({
    'Very Small': gdf['very_small_structures'].mean(),
    'Small':      gdf['small_buildings'].mean(),
    'Medium':     gdf['medium_buildings'].mean(),
    'Large':      gdf['large_buildings'].mean(),
})
avg_mix.plot(kind='bar', ax=axes[0],
             color=['#FFECB3', '#FFE082', '#FFB300', '#E65100'], edgecolor='white')
axes[0].set_title('Average Building Count by Type')
axes[0].set_ylabel('Avg per Settlement')
axes[0].tick_params(axis='x', rotation=0)

axes[1].hist(gdf['Medium_and_large_buildings_pc'].dropna() * 100,
             bins=50, color='#FF9800', edgecolor='white')
axes[1].axvline(10, color='red', linestyle='--', label='10% threshold')
axes[1].axvline(30, color='darkred', linestyle='--', label='30% threshold')
axes[1].set_title('% Medium + Large Buildings')
axes[1].set_xlabel('% of buildings')
axes[1].legend(fontsize=8)

axes[2].hist(
    gdf['Buildings_per_km2'].clip(upper=gdf['Buildings_per_km2'].quantile(0.95)),
    bins=50, color='#795548', edgecolor='white'
)
axes[2].set_title('Building Density (buildings/km2)')
axes[2].set_xlabel('Buildings per km2')

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data/outputs/maps/inv_buildings.png'), dpi=150)
plt.show()

dense    = (gdf['Buildings_per_km2'] > 500).mean() * 100
large_bld = (gdf['Medium_and_large_buildings_pc'] > 0.10).mean() * 100
print(f'Settlements with >500 buildings/km2 : {dense:.1f}%')
print(f'Settlements with >10% medium+large  : {large_bld:.1f}%')
```

## 6. Social Services Distribution

```python
print('=== SOCIAL SERVICES ===')
print(f"Has health facility    : {gdf['has_health_facility'].sum():,} ({gdf['has_health_facility'].mean()*100:.1f}%)")
print(f"Has education facility : {gdf['has_education_facility'].sum():,} ({gdf['has_education_facility'].mean()*100:.1f}%)")
print(f"Has both               : {(gdf['has_health_facility'] & gdf['has_education_facility']).sum():,}")
print(f"Has neither            : {(~gdf['has_health_facility'] & ~gdf['has_education_facility']).sum():,}")

social_dept = gdf.groupby('admin_cgaz_1').agg(
    Pct_health=('has_health_facility', 'mean'),
    Pct_education=('has_education_facility', 'mean'),
).sort_values('Pct_health', ascending=False)
social_dept[['Pct_health', 'Pct_education']] *= 100

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Social Services Distribution', fontsize=13, fontweight='bold')

social_dept[['Pct_health', 'Pct_education']].plot(
    kind='barh', ax=axes[0], color=['#E53935', '#1E88E5'], edgecolor='white'
)
axes[0].set_title('% Settlements with Social Services by Department')
axes[0].set_xlabel('% of Settlements')
axes[0].legend(['Health Facility', 'Education Facility'])

for has_service, color, label in [
    (True,  '#E53935', 'Has health facility'),
    (False, '#90A4AE', 'No health facility'),
]:
    subset = gdf[gdf['has_health_facility'] == has_service]
    axes[1].scatter(
        subset['population'].clip(upper=10000),
        subset['mean_rwi'], alpha=0.2, s=5, color=color, label=label
    )
axes[1].set_xlabel('Population (capped at 10,000)')
axes[1].set_ylabel('Mean RWI')
axes[1].set_title('Health Facilities: Population vs Wealth')
axes[1].legend()

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data/outputs/maps/inv_social_services.png'), dpi=150)
plt.show()

with_s    = gdf[gdf['has_health_facility'] | gdf['has_education_facility']]
without_s = gdf[~gdf['has_health_facility'] & ~gdf['has_education_facility']]
print(f'Avg demand WITH social services    : {with_s["demand_year0_kwh"].mean():,.0f} kWh/yr')
print(f'Avg demand WITHOUT social services : {without_s["demand_year0_kwh"].mean():,.0f} kWh/yr')
print(f'Uplift from anchor loads           : +{(with_s["demand_year0_kwh"].mean()/without_s["demand_year0_kwh"].mean()-1)*100:.1f}%')
```

## 7. Correlation Matrix

```python
corr_cols = [
    'population', 'mean_rwi', 'num_buildings',
    'Medium_and_large_buildings_pc', 'building_density_percent',
    'distance_to_existing_transmission_lines',
    'dist_main_road_km', 'num_connections',
    'demand_year0_kwh', 'mtf_tier'
]
corr_cols = [c for c in corr_cols if c in gdf.columns]
corr_matrix = gdf[corr_cols].corr()

clean_labels = {
    'population': 'Population',
    'mean_rwi': 'RWI (Wealth)',
    'num_buildings': 'Num Buildings',
    'Medium_and_large_buildings_pc': 'Med+Large Bld %',
    'building_density_percent': 'Bld Density %',
    'distance_to_existing_transmission_lines': 'Dist Grid (km)',
    'dist_main_road_km': 'Dist Road (km)',
    'num_connections': 'Num Connections',
    'demand_year0_kwh': 'Demand Y0 (kWh)',
    'mtf_tier': 'MTF Tier',
}
corr_matrix.index   = [clean_labels.get(c, c) for c in corr_matrix.index]
corr_matrix.columns = [clean_labels.get(c, c) for c in corr_matrix.columns]

fig, ax = plt.subplots(figsize=(11, 9))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='RdBu_r',
            center=0, vmin=-1, vmax=1, linewidths=0.5, ax=ax)
ax.set_title('Correlation Matrix - Key Settlement Variables', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data/outputs/maps/inv_correlation_matrix.png'), dpi=150)
plt.show()
```

## 8. Nightlight and Existing Electrification

```python
print('=== NIGHTLIGHT (PROXY FOR ELECTRIFICATION) ===')
print(f"With nightlight    : {gdf['has_nightlight'].sum():,} ({gdf['has_nightlight'].mean()*100:.1f}%)")
print(f"Without nightlight : {(~gdf['has_nightlight']).sum():,} ({(~gdf['has_nightlight']).mean()*100:.1f}%)")

night_dept = gdf.groupby('admin_cgaz_1')['has_nightlight'].mean().sort_values(ascending=False) * 100

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Existing Electrification (Nightlight Proxy)', fontsize=13, fontweight='bold')

night_dept.plot(kind='barh', ax=axes[0], color='#FDD835', edgecolor='gray')
axes[0].axvline(gdf['has_nightlight'].mean() * 100, color='red', linestyle='--',
                label=f'National avg: {gdf["has_nightlight"].mean()*100:.1f}%')
axes[0].set_title('% Settlements with Nightlight by Department')
axes[0].set_xlabel('% of Settlements')
axes[0].legend()

for nl, color, label in [(True, '#FDD835', 'Has nightlight'), (False, '#546E7A', 'No nightlight')]:
    subset = gdf[gdf['has_nightlight'] == nl]
    axes[1].hist(
        subset['distance_to_existing_transmission_lines'].clip(upper=100),
        bins=40, alpha=0.6, color=color, label=label, edgecolor='white'
    )
axes[1].set_title('Grid Distance: Electrified vs Non-Electrified')
axes[1].set_xlabel('Distance to Grid (km)')
axes[1].set_ylabel('Settlements')
axes[1].legend()

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data/outputs/maps/inv_nightlight.png'), dpi=150)
plt.show()

med_lit   = gdf[gdf['has_nightlight']]['distance_to_existing_transmission_lines'].median()
med_unlit = gdf[~gdf['has_nightlight']]['distance_to_existing_transmission_lines'].median()
print(f'Median grid distance - electrified     : {med_lit:.1f} km')
print(f'Median grid distance - non-electrified : {med_unlit:.1f} km')
```

## 9. Summary: Key Findings

```python
dist_col = 'distance_to_existing_transmission_lines'

print('=' * 60)
print('KEY FINDINGS - BENIN SETTLEMENT INVESTIGATION')
print('=' * 60)
print(f'Total settlements : {len(gdf):,}')
print(f'Total population  : {gdf["population"].sum()/1e6:.1f} million')
print(f'Median settlement : {gdf["population"].median():.0f} persons')
print()
print('ELECTRIFICATION STATUS')
print(f'  Electrified (nightlight) : {gdf["has_nightlight"].mean()*100:.1f}%')
print(f'  Not yet electrified      : {(~gdf["has_nightlight"]).mean()*100:.1f}%')
print()
print('GRID PROXIMITY')
print(f'  Within 10km : {(gdf[dist_col]<=10).mean()*100:.1f}% - grid extension viable')
print(f'  Within 25km : {(gdf[dist_col]<=25).mean()*100:.1f}% - grid extension possible')
print(f'  Beyond 25km : {(gdf[dist_col]>25).mean()*100:.1f}% - off-grid needed')
print()
print('DEMAND TIERS')
print(f'  Tier 1 : {(gdf["mtf_tier"]==1).mean()*100:.1f}%')
print(f'  Tier 2 : {(gdf["mtf_tier"]==2).mean()*100:.1f}%')
print(f'  Tier 3+: {(gdf["mtf_tier"]>=3).mean()*100:.1f}%')
print()
print('SOCIAL SERVICES')
print(f'  With health facility    : {gdf["has_health_facility"].mean()*100:.1f}%')
print(f'  With education facility : {gdf["has_education_facility"].mean()*100:.1f}%')
print()
print('Charts saved to data/outputs/maps/')
```
