# 04 — Sensitivity Analysis
**Benin Least-Cost Electrification Analysis**

Tests how technology selection responds to key uncertain parameters identified during NB03.

**Six parameters tested (from `assumptions.yaml → sensitivity_nb04`):**

| # | Parameter | Base case | Range |
|---|---|---|---|
| 1 | Battery replacement year | Year 10 (LFP) | Year 5 vs 10 |
| 2 | MG connection cost | \$280/HH | \$150–\$350 |
| 3 | SHS kit cost | \$150 | \$120–\$180 |
| 4 | Diesel price | \$0.85/L | \$0.75–\$1.00 |
| 5 | Productive use factor | ×1.25/×1.40 | ×1.00–×1.50 |
| 6 | MG demand threshold | 15,000 kWh/yr | 10,000–25,000 |

**Sections:**
1. Load base results
2. Scenario runner function
3. Battery replacement sensitivity
4. MG connection cost sensitivity
5. SHS kit cost sensitivity
6. Diesel price sensitivity
7. MG demand threshold sensitivity
8. Combined tornado chart
9. Save results

```python
PROCESSED_DIR = Path('..') / 'data' / 'processed'
candidates = sorted(PROCESSED_DIR.glob('settlements_lcoe_*.geojson'), reverse=True)
if not candidates:
    raise FileNotFoundError('No settlements_lcoe_*.geojson — run NB03 first')

BASE_PATH = candidates[0]
print(f'Loading: {BASE_PATH.name}')
gdf_base = gpd.read_file(BASE_PATH)
print(f'Loaded {len(gdf_base):,} settlements | {len(gdf_base.columns)} columns')

# Restore demand_timeseries
if 'demand_timeseries' in gdf_base.columns:
    if isinstance(gdf_base['demand_timeseries'].iloc[0], str):
        gdf_base['demand_timeseries'] = gdf_base['demand_timeseries'].apply(ast.literal_eval)

# Base case technology summary
print('\n=== BASE CASE RESULTS ===')
base_summary = gdf_base['least_cost_tech'].value_counts()
for tech, n in base_summary.items():
    print(f'  {tech:<35}: {n:,} ({n/len(gdf_base)*100:.1f}%)')
print(f'\nBase CAPEX: ${gdf_base["least_cost_capex"].sum()/1e6:.1f}M')

TECH_COLORS = {
    'Already Electrified'            : '#BDBDBD',
    'Grid Extension'                 : '#1565C0',
    'Mini-Grid: Solar PV Only'       : '#F9A825',
    'Mini-Grid: Solar-Diesel Hybrid' : '#E53935',
    'Mini-Grid: Mini-Hydro'          : '#00897B',
    'SHS'                            : '#43A047',
}
UNELEC_TECHS = ['Grid Extension','Mini-Grid: Solar PV Only',
                'Mini-Grid: Solar-Diesel Hybrid','Mini-Grid: Mini-Hydro','SHS']
```

## 2. Scenario Runner


```python
import importlib
import src.costs.grid_extension  as ge_mod
import src.costs.mini_grid       as mg_mod
import src.costs.shs             as shs_mod
import src.least_cost.technology_selector as ts_mod

def run_scenario(
    gdf,
    scenario_name,
    # Battery
    battery_replace_yr   = 10,
    # MG connection cost
    mg_solar_conn_cost   = 280,
    mg_hybrid_conn_cost  = 300,
    mg_hydro_conn_cost   = 350,
    # SHS kit cost
    shs_t2_cost          = 150,
    shs_t3_cost          = 350,
    # Diesel price override (None = use DieselPrice column)
    diesel_price_override= None,
    # MG demand thresholds
    mg_solar_min_demand  = 15_000,
    mg_hybrid_min_demand = 25_000,
    mg_hydro_min_demand  = 10_000,
):
    """
    Re-run LCOE pipeline with overridden parameters.
    Returns technology summary DataFrame.
    """
    # Reload modules fresh
    importlib.reload(mg_mod)
    importlib.reload(shs_mod)
    importlib.reload(ts_mod)

    gdf = gdf.copy()
    unelec = gdf['elec_status'] == 'unelectrified'

    # ── Override MG parameters ────────────────────────────────────────────────
    mg_mod.MG_SUBTYPES['mg_solar']['battery_replace_yr']  = battery_replace_yr
    mg_mod.MG_SUBTYPES['mg_hybrid']['battery_replace_yr'] = battery_replace_yr
    mg_mod.MG_SUBTYPES['mg_solar']['capex_per_conn']      = mg_solar_conn_cost
    mg_mod.MG_SUBTYPES['mg_hybrid']['capex_per_conn']     = mg_hybrid_conn_cost
    mg_mod.MG_SUBTYPES['mg_hydro']['capex_per_conn']      = mg_hydro_conn_cost
    mg_mod.MG_SUBTYPES['mg_solar']['min_demand_kwh']      = mg_solar_min_demand
    mg_mod.MG_SUBTYPES['mg_hybrid']['min_demand_kwh']     = mg_hybrid_min_demand
    mg_mod.MG_SUBTYPES['mg_hydro']['min_demand_kwh']      = mg_hydro_min_demand

    # ── Override SHS costs ────────────────────────────────────────────────────
    import src.config as cfg
    cfg.SHS['tier_2']['capex_per_unit'] = shs_t2_cost
    cfg.SHS['tier_3']['capex_per_unit'] = shs_t3_cost

    # ── Override diesel price ─────────────────────────────────────────────────
    if diesel_price_override is not None:
        gdf['_diesel_override'] = diesel_price_override
        orig_col = gdf.get('DieselPrice')
        gdf['DieselPrice'] = diesel_price_override

    # ── Recompute MG LCOE ─────────────────────────────────────────────────────
    with warnings.catch_warnings():
        warnings.simplefilter('ignore')
        gdf_u = mg_mod.add_minigrid_lcoe(gdf[unelec].copy())

    float_mg = ['lcoe_mg_solar','lcoe_mg_hybrid','lcoe_mg_hydro',
                'lcoe_minigrid','minigrid_capex_usd','minigrid_peak_kw']
    str_mg   = ['minigrid_subtype','minigrid_subtype_label']
    bool_mg  = ['minigrid_feasible','mg_hydro_feasible']
    for col in float_mg:
        if col in gdf_u.columns:
            if col not in gdf.columns: gdf[col] = np.nan
            gdf[col] = gdf[col].astype(float)
            gdf.loc[unelec, col] = gdf_u[col].astype(float).values
    for col in str_mg:
        if col in gdf_u.columns:
            if col not in gdf.columns: gdf[col] = ''
            gdf[col] = gdf[col].astype(str)
            gdf.loc[unelec, col] = gdf_u[col].astype(str).values
    for col in bool_mg:
        if col in gdf_u.columns:
            if col not in gdf.columns: gdf[col] = False
            gdf.loc[unelec, col] = gdf_u[col].values
    for col in ['lcoe_mg_solar','lcoe_mg_hybrid','lcoe_mg_hydro','lcoe_minigrid']:
        if col not in gdf.columns: gdf[col] = np.inf
        gdf[col] = gdf[col].astype(float)
        gdf.loc[~unelec, col] = np.inf

    # ── Recompute SHS LCOE ────────────────────────────────────────────────────
    gdf_s = shs_mod.add_shs_lcoe(gdf[unelec].copy())
    for col in ['lcoe_shs','shs_capex_usd']:
        if col in gdf_s.columns:
            if col not in gdf.columns: gdf[col] = np.inf
            gdf[col] = gdf[col].astype(float)
            gdf.loc[unelec, col] = gdf_s[col].astype(float).values
    gdf['lcoe_shs'] = gdf['lcoe_shs'].astype(float)
    gdf.loc[~unelec, 'lcoe_shs'] = np.inf

    # ── Technology selection ──────────────────────────────────────────────────
    importlib.reload(ts_mod)
    from src.least_cost.technology_selector import select_least_cost, technology_summary
    gdf = select_least_cost(gdf)
    summary = technology_summary(gdf)
    summary.insert(0, 'Scenario', scenario_name)

    # Restore diesel price
    if diesel_price_override is not None and orig_col is not None:
        gdf['DieselPrice'] = orig_col

    return summary, gdf

print('Scenario runner ready ✓')
```

## 3. Battery Replacement Sensitivity
Tests LFP (year 10) vs lead-acid (year 5) replacement cycle.
Expected LCOE spread: ~$0.07/kWh for Solar MG.


```python
battery_results = []
for yr, label in [(5,'Lead-acid (yr 5)'),(8,'LFP conservative (yr 8)'),
                   (10,'LFP base case (yr 10)'),(12,'LFP optimistic (yr 12)')]:
    s, _ = run_scenario(gdf_base, label, battery_replace_yr=yr)
    battery_results.append(s)
    print(f'  {label} done')

battery_df = pd.concat(battery_results, ignore_index=True)
print('\n=== BATTERY SENSITIVITY ===')
pivot = battery_df.pivot_table(index='Scenario', columns='Technology',
                                values='Pct_settlements', fill_value=0)
print(pivot[[t for t in UNELEC_TECHS if t in pivot.columns]].round(1).to_string())
```

## 4. MG Connection Cost Sensitivity
Most impactful parameter — shifts settlements between SHS and Solar MG.


```python
conn_results = []
for cost, label in [(150,'Subsidised ($150)'),(200,'Low ($200)'),
                     (280,'Base case ($280)'),(350,'Full cost ($350)')]:
    s, _ = run_scenario(gdf_base, label, mg_solar_conn_cost=cost,
                        mg_hybrid_conn_cost=cost+20, mg_hydro_conn_cost=cost+70)
    conn_results.append(s)
    print(f'  {label} done')

conn_df = pd.concat(conn_results, ignore_index=True)
print('\n=== CONNECTION COST SENSITIVITY ===')
pivot = conn_df.pivot_table(index='Scenario', columns='Technology',
                             values='Pct_settlements', fill_value=0)
print(pivot[[t for t in UNELEC_TECHS if t in pivot.columns]].round(1).to_string())
```

## 5. SHS Kit Cost Sensitivity
Tests market price decline scenario ($120 = projected 2030 price).


```python
shs_results = []
for cost, label in [(120,'Low kit ($120)'),(150,'Base case ($150)'),
                     (180,'High kit ($180)')]:
    s, _ = run_scenario(gdf_base, label, shs_t2_cost=cost)
    shs_results.append(s)
    print(f'  {label} done')

shs_df = pd.concat(shs_results, ignore_index=True)
print('\n=== SHS KIT COST SENSITIVITY ===')
pivot = shs_df.pivot_table(index='Scenario', columns='Technology',
                            values='Pct_settlements', fill_value=0)
print(pivot[[t for t in UNELEC_TECHS if t in pivot.columns]].round(1).to_string())
```

## 6. Diesel Price Sensitivity
Tests impact on Solar-Diesel Hybrid competitiveness.


```python
diesel_results = []
for price, label in [(0.75,'Low ($0.75/L)'),(0.85,'Base ($0.85/L)'),
                      (1.00,'High ($1.00/L)')]:
    s, _ = run_scenario(gdf_base, label, diesel_price_override=price)
    diesel_results.append(s)
    print(f'  {label} done')

diesel_df = pd.concat(diesel_results, ignore_index=True)
print('\n=== DIESEL PRICE SENSITIVITY ===')
pivot = diesel_df.pivot_table(index='Scenario', columns='Technology',
                               values='Pct_settlements', fill_value=0)
print(pivot[[t for t in UNELEC_TECHS if t in pivot.columns]].round(1).to_string())
```

## 7. MG Demand Threshold Sensitivity
Tests impact of ESMAP threshold assumption on technology distribution.


```python
thresh_results = []
for solar_t, label in [
    (10_000, 'Low threshold (10,000 kWh)'),
    (15_000, 'Base case (15,000 kWh)'),
    (25_000, 'High threshold (25,000 kWh)'),
]:
    s, _ = run_scenario(gdf_base, label,
                        mg_solar_min_demand=solar_t,
                        mg_hybrid_min_demand=int(solar_t*1.67),
                        mg_hydro_min_demand=int(solar_t*0.67))
    thresh_results.append(s)
    print(f'  {label} done')

thresh_df = pd.concat(thresh_results, ignore_index=True)
print('\n=== DEMAND THRESHOLD SENSITIVITY ===')
pivot = thresh_df.pivot_table(index='Scenario', columns='Technology',
                               values='Pct_settlements', fill_value=0)
print(pivot[[t for t in UNELEC_TECHS if t in pivot.columns]].round(1).to_string())
```

## 8. Tornado Chart — Parameter Impact
Shows which parameter has the largest impact on Solar MG share.


```python
# Base Solar MG share
base_mg = gdf_base[gdf_base['least_cost_tech']=='Mini-Grid: Solar PV Only'].shape[0] / len(gdf_base) * 100

# Collect min/max impact per parameter
tornado_data = []

def mg_share(df):
    return df[df['Technology']=='Mini-Grid: Solar PV Only']['Pct_settlements'].values

for param, df in [
    ('Connection cost',   conn_df),
    ('SHS kit cost',      shs_df),
    ('Battery lifetime',  battery_df),
    ('Diesel price',      diesel_df),
    ('Demand threshold',  thresh_df),
]:
    vals = []
    for scen in df['Scenario'].unique():
        sub = df[df['Scenario']==scen]
        v = mg_share(sub)
        if len(v): vals.append(float(v[0]))
    if vals:
        tornado_data.append({'Parameter': param,
                             'Min': min(vals), 'Max': max(vals),
                             'Range': max(vals) - min(vals)})

tornado_df = pd.DataFrame(tornado_data).sort_values('Range', ascending=True)

fig, ax = plt.subplots(figsize=(10, 5))
colors_low  = '#E53935'
colors_high = '#43A047'
y_pos = range(len(tornado_df))

ax.barh(list(y_pos), tornado_df['Max'] - base_mg,
        left=base_mg, color=colors_high, alpha=0.8, label='Above base')
ax.barh(list(y_pos), tornado_df['Min'] - base_mg,
        left=base_mg, color=colors_low, alpha=0.8, label='Below base')

ax.axvline(base_mg, color='black', lw=1.5, ls='--', label=f'Base case ({base_mg:.1f}%)')
ax.set_yticks(list(y_pos))
ax.set_yticklabels(tornado_df['Parameter'], fontsize=10)
ax.set_xlabel('Solar MG share (% settlements)')
ax.set_title('Tornado Chart — Solar MG Share Sensitivity\n'
             'Width = range of outcomes across parameter values', fontsize=11)
ax.legend(fontsize=9)

# Annotate range
for i, row in enumerate(tornado_df.itertuples()):
    ax.text(max(row.Max, base_mg) + 0.2, i,
            f'{row.Range:.1f}pp', va='center', fontsize=9)

plt.tight_layout()
plt.savefig('../data/outputs/maps/tornado_chart.png', dpi=150)
plt.show()
print('Tornado chart saved ✓')
```

## 9. Save Results


```python
ts = datetime.now().strftime('%Y%m%d_%H%M')
TABLE_DIR = Path('..') / 'data' / 'outputs' / 'tables'
TABLE_DIR.mkdir(parents=True, exist_ok=True)

all_results = pd.concat([
    battery_df, conn_df, shs_df, diesel_df, thresh_df
], ignore_index=True)
all_results.to_csv(TABLE_DIR / f'sensitivity_results_{ts}.csv', index=False)
tornado_df.to_csv(TABLE_DIR / f'tornado_summary_{ts}.csv', index=False)

print(f'Saved: sensitivity_results_{ts}.csv')
print(f'Saved: tornado_summary_{ts}.csv')
print('\n=== TORNADO SUMMARY ===')
print(tornado_df[['Parameter','Min','Max','Range']].sort_values(
    'Range', ascending=False).to_string(index=False))
```
