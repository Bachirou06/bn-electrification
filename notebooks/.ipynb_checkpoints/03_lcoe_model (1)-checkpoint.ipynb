# 03 — LCOE Model & Least-Cost Technology Selection
**Benin Least-Cost Electrification Analysis**


```python
import sys
sys.path.append('..')
import geopandas as gpd
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import ast, warnings
from pathlib import Path
from datetime import datetime

from src.costs.grid_extension import add_grid_lcoe, add_densification
from src.costs.mini_grid import add_minigrid_lcoe
from src.costs.shs import add_shs_lcoe
from src.least_cost.technology_selector import (
    select_least_cost, add_grid_priority, technology_summary
)
from src.config import GRID, DISCOUNT_RATE, GENERAL
import seaborn as sns
sns.set_theme(style='whitegrid')
print('Libraries loaded ✓')
```

## 1. Load Data


```python
PROCESSED_DIR = Path('..') / 'data' / 'processed'
candidates = sorted(PROCESSED_DIR.glob('settlements_demand_*.geojson'), reverse=True)
if not candidates:
    raise FileNotFoundError('No settlements_demand_*.geojson — run NB02 first')
DEMAND_PATH = candidates[0]
print(f'Loading: {DEMAND_PATH.name}')
gdf = gpd.read_file(DEMAND_PATH)
print(f'Loaded {len(gdf):,} settlements | {len(gdf.columns)} columns')

if 'demand_timeseries' in gdf.columns:
    if isinstance(gdf['demand_timeseries'].iloc[0], str):
        gdf['demand_timeseries'] = gdf['demand_timeseries'].apply(ast.literal_eval)
    print('demand_timeseries deserialised ✓')
else:
    T = GENERAL['planning_horizon_years']
    gdf['demand_timeseries'] = gdf['demand_year0_kwh'].apply(
        lambda d0: [d0 * (1.04)**t for t in range(T+1)])
    print('demand_timeseries reconstructed ✓')

print('\nElectrification status:')
for s, n in gdf['elec_status'].value_counts().items():
    print(f'  {s}: {n:,} ({n/len(gdf)*100:.1f}%)')
```

## 2. Cost Assumptions


```python
TARGET_CONNECTION_RATE = 0.90
LV_COST_PER_KM        = 8_000
LV_LENGTH_PER_HH      = 0.030
CONNECTION_COST       = 400
DENSIF_COST_PER_HH    = LV_COST_PER_KM * LV_LENGTH_PER_HH + CONNECTION_COST

print(f'Grid extension  : slope-adjusted $40k-$55k/km + OnSSET terrain penalty')
print(f'Densification   : ${DENSIF_COST_PER_HH:.0f}/HH (LV ${LV_COST_PER_KM*LV_LENGTH_PER_HH:.0f} + conn ${CONNECTION_COST})')
print(f'Discount rate   : {DISCOUNT_RATE*100:.0f}%')
print(f'Planning horizon: {GENERAL["planning_horizon_years"]} years')
```

## 3. Grid Densification


```python
# VIDA num_connections = satellite rooftops = building count
# connection_rate = 1.0 by definition — no densification gap measurable
gdf['connection_rate']     = 1.0
gdf['HH_to_connect']       = 0
gdf['densification_needed']= False
gdf['densification_capex'] = 0.0
gdf['lcoe_densification']  = np.nan
print(f'Electrified: {(gdf.elec_status=="electrified").sum():,} — already served')
print('Densification not applicable (VIDA has no unconnected-buildings signal)')
```

## 4. Grid Extension LCOE
Slope-adjusted MV cost + OnSSET terrain penalty. Runs on unelectrified only.


```python
unelec_mask = gdf['elec_status'] == 'unelectrified'
print(f'Grid extension LCOE for {unelec_mask.sum():,} unelectrified settlements...')
gdf_unelec = gdf[unelec_mask].copy()
gdf_unelec = add_grid_lcoe(gdf_unelec, dist_col='GridDistKm')

for col in ['grid_penalty','mv_cost_per_km','grid_capex_usd','lcoe_grid','grid_feasible']:
    if col in gdf_unelec.columns:
        gdf.loc[unelec_mask, col] = gdf_unelec[col].values

gdf.loc[~unelec_mask, 'lcoe_grid']      = 0.0
gdf.loc[~unelec_mask, 'grid_feasible']  = True
gdf.loc[~unelec_mask, 'grid_capex_usd'] = 0.0

fin = gdf.loc[unelec_mask,'lcoe_grid'].replace(np.inf,np.nan).dropna()
print(f'Feasible: {gdf.loc[unelec_mask,"grid_feasible"].sum():,} | Median LCOE: ${fin.median():.3f}/kWh')
```

## 5. Mini-Grid LCOE


```python
print(f'Mini-grid LCOE for {unelec_mask.sum():,} unelectrified settlements...')
gdf_mg = gdf[unelec_mask].copy()
with warnings.catch_warnings(record=True) as w:
    warnings.simplefilter('always')
    gdf_mg = add_minigrid_lcoe(gdf_mg)
    for warning in w:
        print(f'  ⚠ {warning.message}')

mg_cols = ['lcoe_mg_solar','lcoe_mg_hybrid','lcoe_mg_hydro',
           'lcoe_minigrid','minigrid_subtype','minigrid_subtype_label',
           'minigrid_capex_usd','minigrid_peak_kw','minigrid_feasible','mg_hydro_feasible']
for col in mg_cols:
    if col in gdf_mg.columns:
        if col not in gdf.columns: gdf[col] = np.inf
        gdf.loc[unelec_mask, col] = gdf_mg[col].values
for col in ['lcoe_mg_solar','lcoe_mg_hybrid','lcoe_mg_hydro','lcoe_minigrid']:
    if col not in gdf.columns: gdf[col] = np.inf
    gdf.loc[~unelec_mask, col] = np.inf

for col in ['lcoe_mg_solar','lcoe_mg_hybrid','lcoe_mg_hydro']:
    fin  = gdf.loc[unelec_mask, col].replace(np.inf,np.nan).dropna()
    feas = (gdf.loc[unelec_mask, col] < np.inf).sum()
    med  = f'{fin.median():.3f}' if len(fin) > 0 else 'n/a'
    print(f'  {col:<20}: feasible={feas:,}  median=${med}')
```

## 6. SHS LCOE


```python
print(f'SHS LCOE for {unelec_mask.sum():,} unelectrified settlements...')
gdf_shs = gdf[unelec_mask].copy()
gdf_shs = add_shs_lcoe(gdf_shs)
for col in ['lcoe_shs','shs_capex_usd','shs_kit','shs_feasible']:
    if col in gdf_shs.columns:
        if col not in gdf.columns: gdf[col] = np.nan
        gdf.loc[unelec_mask, col] = gdf_shs[col].values
if 'lcoe_shs' not in gdf.columns: gdf['lcoe_shs'] = np.inf
gdf.loc[~unelec_mask, 'lcoe_shs'] = np.inf

print('\n=== LCOE SUMMARY (unelectrified) ===')
for col, lbl in [('lcoe_grid','Grid'),('lcoe_mg_solar','MG Solar'),
                  ('lcoe_mg_hybrid','MG Hybrid'),('lcoe_mg_hydro','MG Hydro'),('lcoe_shs','SHS')]:
    if col in gdf.columns:
        vals = gdf.loc[unelec_mask, col].replace(np.inf,np.nan).dropna()
        feas = (gdf.loc[unelec_mask, col] < np.inf).sum()
        med  = f'{vals.median():.3f}' if len(vals) > 0 else 'n/a'
        print(f'  {lbl:<12}: median=${med}  feasible={feas:,} ({feas/unelec_mask.sum()*100:.0f}%)')
```

## 7. Least-Cost Selection


```python
gdf = select_least_cost(gdf)
gdf = add_grid_priority(gdf)
summary = technology_summary(gdf)
print('=== LEAST-COST RESULTS ===')
print(summary.to_string(index=False))
print(f'\nTotal CAPEX: ${gdf.loc[unelec_mask,"least_cost_capex"].sum()/1e9:.2f}B')
shs_mask = gdf['least_cost_tech'] == 'SHS'
if shs_mask.sum() > 0:
    print(f'\nSHS kit: {gdf.loc[shs_mask,"shs_kit"].value_counts().to_dict()}')
```

## 8. Visualisation


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('LCOE Comparison — Unelectrified Settlements', fontsize=13, fontweight='bold')
unelec = gdf[gdf['elec_status'] == 'unelectrified']
cols   = ['lcoe_grid','lcoe_mg_solar','lcoe_mg_hybrid','lcoe_mg_hydro','lcoe_shs']
labels = ['Grid','MG Solar','MG Hybrid','MG Hydro','SHS']
colors = ['#1565C0','#FDD835','#E53935','#0288D1','#43A047']
data   = [unelec[c].replace(np.inf,np.nan).dropna().clip(upper=2.0).values
          for c in cols if c in unelec.columns]
valid_labels = [l for l,c in zip(labels,cols) if c in unelec.columns]
bp = axes[0].boxplot(data, labels=valid_labels, patch_artist=True)
for patch, color in zip(bp['boxes'], colors): patch.set_facecolor(color); patch.set_alpha(0.7)
axes[0].set_ylabel('LCOE (USD/kWh, clipped $2)'); axes[0].set_title('LCOE by Technology')
axes[0].tick_params(axis='x', rotation=25)

tech_counts = gdf['least_cost_tech'].value_counts()
tc_map = {'Already Electrified':'#90CAF9','Grid Extension':'#1565C0',
          'Mini-Grid: Solar PV Only':'#FDD835','Mini-Grid: Solar-Diesel Hybrid':'#E53935',
          'Mini-Grid: Mini-Hydro':'#0288D1','SHS':'#43A047'}
bars = axes[1].bar(range(len(tech_counts)), tech_counts.values,
                   color=[tc_map.get(t,'#999') for t in tech_counts.index], edgecolor='white')
axes[1].set_xticks(range(len(tech_counts)))
axes[1].set_xticklabels([t.replace('Mini-Grid: ','MG\n') for t in tech_counts.index], fontsize=7, rotation=20)
axes[1].set_ylabel('Settlements'); axes[1].set_title('Technology Selection')
for bar,(t,n) in zip(bars,tech_counts.items()):
    axes[1].text(bar.get_x()+bar.get_width()/2, bar.get_height()+20,
                 f'{n:,}\n({n/len(gdf)*100:.1f}%)', ha='center', fontsize=7)

win = gdf.loc[unelec_mask,'least_cost_lcoe'].replace(np.inf,np.nan).dropna()
axes[2].hist(win.clip(upper=win.quantile(0.95)), bins=60, color='#7B1FA2', edgecolor='white')
axes[2].axvline(win.median(), color='red', ls='--', label=f'Median=${win.median():.3f}')
axes[2].set_xlabel('LCOE (USD/kWh)'); axes[2].set_ylabel('Settlements')
axes[2].set_title('Winning LCOE Distribution'); axes[2].legend(fontsize=8)

plt.tight_layout()
plt.savefig('../data/outputs/maps/lcoe_comparison.png', dpi=150)
plt.show()
```

## 9. Technology Map


```python
TECH_COLORS = {'Already Electrified':'#90CAF9','Grid Extension':'#1565C0',
               'Mini-Grid: Solar PV Only':'#FDD835','Mini-Grid: Solar-Diesel Hybrid':'#E53935',
               'Mini-Grid: Mini-Hydro':'#0288D1','SHS':'#43A047'}
TECH_SIZES  = {'Already Electrified':3,'Grid Extension':6,'Mini-Grid: Solar PV Only':4,
               'Mini-Grid: Solar-Diesel Hybrid':4,'Mini-Grid: Mini-Hydro':5,'SHS':2}
fig, ax = plt.subplots(figsize=(9,11))
ax.set_facecolor('#E8F4F8')
for bp in [Path('..')/'data'/'raw'/'Benin_boundary.gpkg',
           Path('..')/'data'/'raw'/'benin_boundary.gpkg']:
    if bp.exists():
        gpd.read_file(bp).plot(ax=ax,color='white',edgecolor='#333',linewidth=1.2,zorder=1); break
tp = Path('..')/'data'/'raw'/'Benin_existing_transmission_lines_2017.geojson'
if tp.exists():
    gpd.read_file(tp).plot(ax=ax,color='#212121',linewidth=0.8,alpha=0.6,zorder=2)
for tech in ['SHS','Mini-Grid: Solar PV Only','Mini-Grid: Solar-Diesel Hybrid',
             'Mini-Grid: Mini-Hydro','Grid Extension','Already Electrified']:
    sub = gdf[gdf['least_cost_tech']==tech]
    if len(sub): sub.plot(ax=ax,color=TECH_COLORS.get(tech,'#999'),
                          markersize=TECH_SIZES.get(tech,3),alpha=0.7,marker='o',linewidth=0,zorder=3)
ax.set_xlim(0.8,3.9); ax.set_ylim(6.1,12.5)
counts = gdf['least_cost_tech'].value_counts()
leg = [mpatches.Patch(color=TECH_COLORS.get(t,'#999'),
       label=f'{t} ({counts.get(t,0):,} — {counts.get(t,0)/len(gdf)*100:.1f}%)')
       for t in TECH_COLORS if counts.get(t,0)>0]
ax.legend(handles=leg, loc='lower left', fontsize=7, framealpha=0.95)
ax.set_title(f'Benin — Least-Cost Technology\n17,205 settlements', fontsize=11, fontweight='bold')
ax.set_xlabel('Longitude'); ax.set_ylabel('Latitude')
plt.tight_layout()
plt.savefig('../data/outputs/maps/technology_map.png', dpi=150, bbox_inches='tight')
plt.show(); print('Map saved ✓')
```

## 10. Save Results


```python
ts = datetime.now().strftime('%Y%m%d_%H%M')
OUT_DIR   = Path('..') / 'data' / 'processed'
TABLE_DIR = Path('..') / 'data' / 'outputs' / 'tables'
TABLE_DIR.mkdir(parents=True, exist_ok=True)

save_cols = [c for c in gdf.columns if c != 'demand_timeseries']
gdf[save_cols].to_file(OUT_DIR / f'settlements_lcoe_{ts}.geojson', driver='GeoJSON')
summary = technology_summary(gdf)
summary.to_csv(TABLE_DIR / f'technology_summary_{ts}.csv', index=False)
print(f'Saved: settlements_lcoe_{ts}.geojson')
print(f'Saved: technology_summary_{ts}.csv')
print('\n=== FINAL SUMMARY ===')
print(summary.to_string(index=False))
```
