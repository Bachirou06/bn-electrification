## 2. Modification 1 — Dynamic MTF Tier Assignment

**What OnSSET does by default:**
OnSSET accepts `ResidentialDemandTierTarget` as a fixed scalar input — the same tier applied to ALL settlements nationwide. This is a simplification that ignores local wealth and infrastructure signals.

**Our modification:**
We dynamically assign tiers per settlement using three proxy signals available in the VIDA dataset:
- `mean_rwi` — Relative Wealth Index
- `Medium_and_large_buildings_pc` — building size mix
- `has_health_facility` / `has_education_facility` — social services

This produces a spatially differentiated demand surface instead of a uniform national tier.

````python
# ── MODIFICATION 1: Dynamic MTF Tier Assignment ────────────────────────────────
# Replaces OnSSET's fixed ResidentialDemandTierTarget scalar
# with a per-settlement tier based on wealth + building + services signals

def assign_dynamic_tier(row: pd.Series) -> int:
    """
    Assign MTF demand tier (1-4) per settlement.

    Scoring:
      mean_rwi >= 0.5                        → +2 pts
      mean_rwi >= 0.0                        → +1 pt
      Medium_and_large_buildings_pc >= 0.30  → +2 pts
      Medium_and_large_buildings_pc >= 0.10  → +1 pt
      has_health or has_education facility   → +1 pt
    ─────────────────────────────────────────────────
      Score 0-1 → Tier 1  (4.3 kWh/HH/year)
      Score 2   → Tier 2  (73 kWh/HH/year)
      Score 3   → Tier 3  (365 kWh/HH/year)
      Score 4-5 → Tier 4  (1095 kWh/HH/year)
    """
    score = 0

    # Wealth signal
    rwi = row.get('mean_rwi', np.nan)
    if pd.notna(rwi):
        if rwi >= 0.5:  score += 2
        elif rwi >= 0:  score += 1

    # Building mix signal
    bld = row.get('Medium_and_large_buildings_pc', np.nan)
    if pd.notna(bld):
        if bld >= 0.30:  score += 2
        elif bld >= 0.10: score += 1

    # Social services signal
    has_h = bool(row.get('has_health_facility', False))
    has_e = bool(row.get('has_education_facility', False))
    if has_h or has_e:
        score += 1

    # Map score to tier
    if score <= 1: return 1
    elif score == 2: return 2
    elif score == 3: return 3
    else: return 4


# Apply to all settlements
gdf['ResidentialDemandTierTarget'] = gdf.apply(assign_dynamic_tier, axis=1)

# Compare with OnSSET default (fixed Tier 2 for all)
gdf['OnSSET_default_tier'] = 2   # OnSSET would use this for all settlements

print('=== MODIFICATION 1: Tier Assignment Comparison ===')
print(f"OnSSET default    : Tier 2 for all {len(gdf):,} settlements")
print(f"Our modification  : Dynamic assignment")
print()
print('Dynamic tier distribution:')
tier_counts = gdf['ResidentialDemandTierTarget'].value_counts().sort_index()
for tier, count in tier_counts.items():
    kwh = {1: 4.3, 2: 73, 3: 365, 4: 1095}[tier]
    print(f'  Tier {tier} ({kwh:>6} kWh/HH/yr): {count:,} settlements ({count/len(gdf)*100:.1f}%)')

# Impact on total demand
tier_kwh = {1: 4.3, 2: 73.0, 3: 365.0, 4: 1095.0}
hh_size = GENERAL['avg_household_size']

demand_dynamic = sum(
    (row['population'] / hh_size) * tier_kwh[row['ResidentialDemandTierTarget']]
    for _, row in gdf.iterrows()
) / 1e6

demand_default = sum(
    (row['population'] / hh_size) * tier_kwh[2]
    for _, row in gdf.iterrows()
) / 1e6

print(f'\nTotal demand — OnSSET default (Tier 2 all) : {demand_default:.1f} GWh/year')
print(f'Total demand — Our dynamic assignment      : {demand_dynamic:.1f} GWh/year')
print(f'Difference                                 : {demand_dynamic - demand_default:+.1f} GWh/year ({(demand_dynamic/demand_default - 1)*100:+.1f}%)')
````

````python
# Visualize tier spatial distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Modification 1: Dynamic vs Fixed Tier Assignment', fontsize=13, fontweight='bold')

tier_colors = {1: '#FFF9C4', 2: '#FFE082', 3: '#FFB300', 4: '#E65100'}

# Dynamic assignment
for tier in [1, 2, 3, 4]:
    subset = gdf[gdf['ResidentialDemandTierTarget'] == tier]
    axes[0].scatter(
        subset.geometry.centroid.x,
        subset.geometry.centroid.y,
        c=tier_colors[tier], s=1, alpha=0.5, label=f'Tier {tier}'
    )
axes[0].set_title('Our Modification: Dynamic Tier Assignment')
axes[0].legend(markerscale=5)
axes[0].set_xlabel('Longitude')
axes[0].set_ylabel('Latitude')

# OnSSET default (all Tier 2)
axes[1].scatter(
    gdf.geometry.centroid.x,
    gdf.geometry.centroid.y,
    c=tier_colors[2], s=1, alpha=0.5, label='Tier 2 (all)'
)
axes[1].set_title('OnSSET Default: Fixed Tier 2 for All Settlements')
axes[1].legend(markerscale=5)
axes[1].set_xlabel('Longitude')

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data' / 'outputs' / 'maps' / 'mod1_tier_comparison.png'), dpi=150)
plt.show()
````

## 3. Modification 2 — Spatially Explicit Solar Resource

**What OnSSET does by default:**
OnSSET uses a single GHI value per settlement from a GIS raster layer, but applies a **fixed national capacity factor** for mini-grid sizing. This ignores the spatial variation in solar resource across Benin.

**Our modification:**
We assign Peak Sun Hours (PSH) based on latitude zones, then use this to size the PV system correctly:
```
peak_kW = annual_demand_kWh / (PSH × 365 × (1 - system_losses))
```
This replaces the flat `peak_demand_factor = 0.25 kW/HH` with a demand-driven, resource-aware sizing.

````python
# ── MODIFICATION 2: Spatially Explicit Solar Resource ─────────────────────────
# Replaces OnSSET's fixed capacity factor with latitude-based PSH zones

# Solar resource zones for Benin (kWh/m²/day)
# Source: Global Solar Atlas — Benin average GHI by latitude band
SOLAR_ZONES = {
    'north':   {'lat_min': 9.0,  'psh': 5.5,  'label': 'North Benin (Alibori, Atacora)'},
    'central': {'lat_min': 7.0,  'psh': 5.0,  'label': 'Central Benin (Borgou, Collines)'},
    'south':   {'lat_min': 0.0,  'psh': 4.5,  'label': 'South Benin (Atlantique, Littoral)'},
}

SYSTEM_LOSSES  = 0.20   # 20% total system losses (wiring, inverter, soiling)
PANEL_EFF      = 0.18   # 18% monocrystalline panel efficiency
BATTERY_HOURS  = 4      # hours of battery storage

def assign_psh(lat: float) -> float:
    """Assign Peak Sun Hours based on latitude zone."""
    if lat >= 9.0:  return 5.5   # North
    elif lat >= 7.0: return 5.0  # Central
    else:            return 4.5  # South


def size_pv_system(annual_demand_kwh: float, psh: float) -> dict:
    """
    Size a PV mini-grid system based on actual demand and solar resource.

    Formula:
      peak_kW = annual_demand_kWh / (PSH × 365 × (1 - losses))
      battery_kWh = peak_kW × battery_hours

    This replaces OnSSET's fixed peak_demand_factor = 0.25 kW/HH
    with a demand-driven, resource-aware approach.
    """
    peak_kw     = annual_demand_kwh / (psh * 365 * (1 - SYSTEM_LOSSES))
    battery_kwh = peak_kw * BATTERY_HOURS
    return {
        'pv_peak_kw':    round(peak_kw, 3),
        'battery_kwh':   round(battery_kwh, 3),
        'psh':           psh,
        'capacity_factor': psh / 24,   # effective capacity factor
    }


# Apply to all settlements
gdf['psh'] = gdf['lat'].apply(assign_psh)

# Size PV system using demand from Notebook 01
pv_sizing = gdf.apply(
    lambda r: size_pv_system(r['demand_year0_kwh'], r['psh']),
    axis=1,
    result_type='expand'
)
gdf['pv_peak_kw']    = pv_sizing['pv_peak_kw']
gdf['battery_kwh']   = pv_sizing['battery_kwh']
gdf['capacity_factor'] = pv_sizing['capacity_factor']

# Compare with OnSSET default sizing (fixed 0.25 kW/HH)
gdf['onsset_default_peak_kw'] = gdf['num_households'] * 0.25

print('=== MODIFICATION 2: Solar Resource Zones ===')
for zone, info in SOLAR_ZONES.items():
    n = (gdf['psh'] == info['psh']).sum()
    print(f"  {info['label']:40s}: PSH={info['psh']} kWh/m²/day | {n:,} settlements")

print()
print('PV system sizing comparison:')
print(f"  OnSSET default (0.25 kW/HH) — total PV capacity : {gdf['onsset_default_peak_kw'].sum()/1000:.1f} MW")
print(f"  Our modification (demand-driven) — total PV cap  : {gdf['pv_peak_kw'].sum()/1000:.1f} MW")
print()
print('Per-settlement PV sizing stats (our modification):')
print(gdf[['pv_peak_kw', 'battery_kwh', 'psh']].describe().round(2))
````

````python
# Visualize solar zones and sizing difference
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Modification 2: Spatially Explicit Solar Resource', fontsize=13, fontweight='bold')

zone_colors = {4.5: '#FFF176', 5.0: '#FFB300', 5.5: '#E65100'}
zone_labels = {4.5: 'South (PSH=4.5)', 5.0: 'Central (PSH=5.0)', 5.5: 'North (PSH=5.5)'}

for psh_val, color in zone_colors.items():
    subset = gdf[gdf['psh'] == psh_val]
    axes[0].scatter(
        subset.geometry.centroid.x,
        subset.geometry.centroid.y,
        c=color, s=1, alpha=0.5, label=zone_labels[psh_val]
    )
axes[0].set_title('Solar Resource Zones (PSH)')
axes[0].legend(markerscale=5)
axes[0].set_xlabel('Longitude')
axes[0].set_ylabel('Latitude')

# Sizing comparison scatter
axes[1].scatter(
    gdf['onsset_default_peak_kw'].clip(upper=500),
    gdf['pv_peak_kw'].clip(upper=500),
    alpha=0.1, s=5, c=gdf['psh'],
    cmap='YlOrRd'
)
lim = 500
axes[1].plot([0, lim], [0, lim], 'r--', alpha=0.5, label='1:1 line')
axes[1].set_xlabel('OnSSET default sizing (kW) — fixed 0.25 kW/HH')
axes[1].set_ylabel('Our modification (kW) — demand-driven')
axes[1].set_title('PV System Sizing Comparison')
axes[1].legend()

plt.tight_layout()
plt.savefig(str(BASE_DIR / 'data' / 'outputs' / 'maps' / 'mod2_solar_resource.png'), dpi=150)
plt.show()
````

## 4. Build OnSSET Input CSV

Map all settlement attributes to OnSSET column names.
Reference: OnSSET column specification from `onsset/onsset.py`

````python
# ── OnSSET Column Mapping ──────────────────────────────────────────────────────
#
# OnSSET column          ← Our settlement attribute
# ─────────────────────────────────────────────────
# X                      ← lon
# Y                      ← lat
# Pop                    ← population
# NightLights            ← has_nightlight
# TravelHours            ← dist_main_road_km (proxy)
# GHI                    ← assigned from psh zones
# WindVel                ← 0 (not available)
# GridDistCurrent        ← distance_to_existing_transmission_lines
# GridDistPlanned        ← distance_to_planned_transmission_lines
# SubstationDist         ← dist_to_substations
# RoadDist               ← dist_main_road_km
# ResidentialDemandTierTarget ← ResidentialDemandTierTarget (our Mod 1)
# IsUrban                ← derived from building_density_percent
# HealthDist             ← has_health_facility (binary proxy)
# EducationDist          ← has_education_facility (binary proxy)

def build_onsset_csv(gdf: gpd.GeoDataFrame) -> pd.DataFrame:
    """
    Convert settlement GeoDataFrame to OnSSET-compatible input CSV.
    Applies both Modification 1 (dynamic tiers) and Modification 2 (solar).
    """
    df = pd.DataFrame()

    # ── Location ──────────────────────────────────────────────────────────────
    df['X']   = gdf['lon']
    df['Y']   = gdf['lat']

    # ── Population ────────────────────────────────────────────────────────────
    df['Pop'] = gdf['population']

    # ── Electrification status ────────────────────────────────────────────────
    # 1 = electrified, 0 = not electrified
    # Use has_nightlight as proxy for current electrification
    df['ElecPopCalib'] = gdf['has_nightlight'].fillna(0).astype(int)

    # ── Grid infrastructure ───────────────────────────────────────────────────
    df['GridDistCurrent'] = gdf['distance_to_existing_transmission_lines'].fillna(9999)
    df['GridDistPlanned'] = gdf.get(
        'distance_to_planned_transmission_lines',
        gdf['distance_to_existing_transmission_lines']
    ).fillna(9999)
    df['SubstationDist']  = gdf['dist_to_substations'].fillna(9999)
    df['RoadDist']        = gdf['dist_main_road_km'].fillna(50)

    # ── Solar resource (Modification 2) ───────────────────────────────────────
    df['GHI']             = gdf['psh']   # Peak Sun Hours as GHI proxy

    # ── Wind (not available — set to 0) ───────────────────────────────────────
    df['WindVel']         = 0.0

    # ── Demand tier (Modification 1) ──────────────────────────────────────────
    df['ResidentialDemandTierTarget'] = gdf['ResidentialDemandTierTarget']

    # ── Urban classification ──────────────────────────────────────────────────
    # OnSSET uses IsUrban to apply different demand growth rates
    df['IsUrban'] = (gdf['building_density_percent'].fillna(0) > 30).astype(int)

    # ── Social services ───────────────────────────────────────────────────────
    df['HealthFacilities']    = gdf['num_health_facilities'].fillna(0).astype(int)
    df['EducationFacilities'] = gdf['num_education_facilities'].fillna(0).astype(int)

    # ── Additional identifiers ────────────────────────────────────────────────
    df['id']           = gdf['identifier']
    df['village_name'] = gdf['village_name']
    df['admin1']       = gdf['admin_cgaz_1']
    df['admin2']       = gdf['admin_cgaz_2']

    # ── Our pre-computed demand (for cross-validation) ─────────────────────────
    df['our_demand_year0_kwh'] = gdf['demand_year0_kwh']
    df['our_mtf_tier']         = gdf['mtf_tier']
    df['pv_peak_kw_mod2']      = gdf['pv_peak_kw']

    return df


onsset_df = build_onsset_csv(gdf)
print(f'OnSSET input CSV: {len(onsset_df):,} rows × {len(onsset_df.columns)} columns')
print()
print('Column summary:')
print(onsset_df.describe().round(2))
````

````python
# Quality checks before saving
print('=== QUALITY CHECKS ===')
print(f"Rows with missing GridDistCurrent : {(onsset_df['GridDistCurrent'] == 9999).sum():,}")
print(f"Rows with zero population         : {(onsset_df['Pop'] == 0).sum():,}")
print(f"Urban settlements                 : {onsset_df['IsUrban'].sum():,} ({onsset_df['IsUrban'].mean()*100:.1f}%)")
print(f"Currently electrified             : {onsset_df['ElecPopCalib'].sum():,} ({onsset_df['ElecPopCalib'].mean()*100:.1f}%)")
print()
print('Demand tier distribution:')
print(onsset_df['ResidentialDemandTierTarget'].value_counts().sort_index())
print()
print('GHI / PSH zones:')
print(onsset_df['GHI'].value_counts().sort_index())
````

````python
# Save OnSSET input CSV
output_path = BASE_DIR / 'data' / 'processed' / 'benin_onsset_input.csv'
onsset_df.to_csv(str(output_path), index=False)
print(f'OnSSET input CSV saved → {output_path}')
print()
print('Next steps:')
print('  1. Copy benin_onsset_input.csv to the OnSSET repository folder')
print('  2. Run 1. OnSSET_Calibration.ipynb')
print('  3. Run 2. OnSSET_Scenarios.ipynb')
print('  4. Load results back here for cross-validation (Section 5 below)')
````

## 5. Pseudo-Code: How OnSSET Uses Our Modifications

This section shows exactly where our modifications plug into OnSSET's codebase — useful for the technical presentation.

````python
# ── PSEUDO-CODE: OnSSET Calibration Notebook modifications ────────────────────
# File: 1. OnSSET_Calibration.ipynb
# ─────────────────────────────────────────────────────────────────────────────

onsset_calibration_pseudocode = """
# ── ORIGINAL OnSSET (fixed tier for all settlements) ──────────────────────────
scenario_parameters = {
    'start_year': 2025,
    'end_year': 2030,
    'tier_of_access': 2,          # ← FIXED: same tier for entire country
    'rural_tier': 2,
    'urban_tier': 2,
}

# ── OUR MODIFICATION 1: Dynamic tier per settlement ───────────────────────────
# Instead of fixed tier_of_access = 2, we pass the pre-computed column:
#   ResidentialDemandTierTarget  ← assigned per settlement using RWI + buildings
#
# In OnSSET's SettlementProcessor.calculate_total_demand_per_settlement():
#
#   ORIGINAL:
#     tier_demand = TIER_KWH[scenario_parameters['tier_of_access']]
#     settlement['demand'] = num_hh × tier_demand × growth_rate^t
#
#   MODIFIED:
#     tier_demand = TIER_KWH[settlement['ResidentialDemandTierTarget']]  ← per row
#     settlement['demand'] = num_hh × tier_demand × growth_rate^t
"""

onsset_scenarios_pseudocode = """
# ── ORIGINAL OnSSET (fixed capacity factor for mini-grid sizing) ──────────────
# In Technology class for PV mini-grid:
#   peak_kw = num_households × 0.25   # fixed kW/HH regardless of location

# ── OUR MODIFICATION 2: Demand-driven, solar-aware sizing ─────────────────────
# In Technology.calculate_lcoe() for mini-grid:
#
#   ORIGINAL:
#     peak_kw = num_hh × peak_demand_factor  # flat 0.25 kW/HH
#
#   MODIFIED:
#     psh = assign_psh(settlement['Y'])      # latitude-based PSH
#     peak_kw = annual_demand_kwh / (psh × 365 × (1 - system_losses))
#
# Effect:
#   North Benin (PSH=5.5): smaller PV system needed → lower LCOE → mini-grid more competitive
#   South Benin (PSH=4.5): larger PV system needed  → higher LCOE → SHS more competitive
"""

print('=== MODIFICATION 1 — Where it plugs into OnSSET ===')
print(onsset_calibration_pseudocode)
print('=== MODIFICATION 2 — Where it plugs into OnSSET ===')
print(onsset_scenarios_pseudocode)
````
