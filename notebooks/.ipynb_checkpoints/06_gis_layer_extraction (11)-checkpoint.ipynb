# Notebook 06 — GIS Layer Extraction & Settlement Enrichment

## Data Source: VIDA / DRE Atlas Settlement Data

This notebook enriches the Benin settlement dataset produced by **VIDA** for the
**DRE Atlas Planning Platform**, commissioned by ESMAP, the World Bank, and IFC.

> *VIDA, Fischer M. & La Costa J. (2024). DRE Atlas Planning Platform — Technical
> Documentation. Prepared for ESMAP / World Bank Group.*  
> Platform: [https://energydata.info](https://energydata.info) (search "DRE Benin")  
> Contact: malin@vida.place, jess@vida.place

### What the VIDA/DRE Atlas data contains (per settlement)
The dataset provides **60+ settlement-level characteristics** extracted from satellite
imagery and geospatial analysis across 58 countries, including Benin (17,205 settlements).

#### Settlement identification methodology (VIDA Technical Doc, Section 2)
- Building footprints from **Google + Microsoft** open datasets (merged global coverage),
  detected from very-high-resolution (30 cm) satellite imagery
- **DBScan clustering** algorithm applied to identify settlement boundaries (alpha shapes)
- Parameters: min. building size 15 m², max. inter-building distance 150 m,
  min. 20 neighbours, min. 4 buildings per settlement

#### Building size classification (VIDA Technical Doc, Annex)
| Category | Footprint size | Role in demand model |
|---|---|---|
| `large_buildings` | > 200 m² | 1,100 Wh/day |
| `medium_buildings` | 120–200 m² | 650 Wh/day |
| `small_buildings` | 15–120 m² | 10%×650 + 90%×300 Wh/day |
| `very_small_structures` | < 15 m² | **Excluded** (uninhabited: sheds, kitchens) |

#### Official VIDA energy demand formula (VIDA Technical Doc, p.11)
```
energy_demand (kWh/day) = large_buildings × 1.100
                        + medium_buildings × 0.650
                        + small_buildings × (0.10 × 0.650 + 0.90 × 0.300)
                        = large × 1.100 + medium × 0.650 + small × 0.2170
```
This formula is validated at **r = 1.000, MAE = 0.000** against the `energy_demand`
column in the Benin dataset — confirming exact implementation of the VIDA methodology.

#### Key columns used in this notebook and their official VIDA definitions
| Column | Unit | VIDA Definition |
|---|---|---|
| `energy_demand` | kWh/day | Total settlement demand (all eligible buildings) |
| `avg_connection_energy_demand` | kWh/day | `energy_demand / num_connections` |
| `num_connections` | count | = `num_buildings` (all buildings ≥ 15 m²) |
| `distance_to_existing_transmission_lines` | km | Distance to closest existing MV/HV line |
| `distance_to_planned_transmission_lines` | km | Distance to closest planned MV/HV line |
| `dist_to_substations` | km | Distance to nearest substation |
| `dist_nearest_hub_km` | km | Distance to nearest **regional hub** (city/town) — **NOT grid distance** |
| `dist_lake_river_km` | km | Distance to closest river or lake (HydroSHEDS) |
| `dist_main_road_km` | km | Distance to closest main road (OSM motorway/trunk/primary/secondary) |
| `travel_time_min` | min | Travel time to nearest city |
| `mean_rwi` | index | Meta Relative Wealth Index (mean within settlement) |
| `has_nightlight` | bool | VIDA GridLight algorithm on NASA VIIRS (Sep 2022–Aug 2023) |

#### Data source for grid distances (VIDA Technical Doc, p.12 & Annex)
Grid infrastructure distances are computed from **World Bank energydata.info** grid data
(MV, HV lines — existing and planned). VIDA notes that *"publicly available information on
the existing and planned transmission grid can be incomplete and/or outdated"* — this
uncertainty is propagated through the LCOE sensitivity analysis in Notebook 07.

#### `dist_nearest_hub_km` clarification (VIDA Technical Doc, p.13 & Annex)
> *"Distance to the closest regional hub (populated places of types: city, town)"* — OpenStreetMap  

This is a **socio-economic accessibility indicator**, not an electricity infrastructure distance.
It is used in this pipeline as `CityDistKm` (demand signal / market access proxy) and is
**not** used as the grid extension distance in LCOE calculations.

## Shared utilities

````python
# ── Shared raster extraction utility ──────────────────────────────────────────

def raster_diagnostics(path: Path, label: str):
    """Print raster metadata and value range — always run before extracting."""
    with rasterio.open(path) as src:
        band  = src.read(1).astype(float)
        nd    = src.nodata
        if nd is not None:
            valid = band[band != nd]
        else:
            valid = band.flatten()
        valid = valid[np.isfinite(valid)]
        print(f"  {label} raster:")
        print(f"    File    : {path.name}")
        print(f"    Size    : {src.width} × {src.height} px  |  CRS: {src.crs}")
        print(f"    NoData  : {nd}")
        print(f"    Range   : {valid.min():.4f}  –  {valid.max():.4f}")
        print(f"    Mean    : {valid.mean():.4f}  |  Std: {valid.std():.4f}")
        print(f"    Valid px: {len(valid):,} / {band.size:,} ({len(valid)/band.size*100:.1f}%)")
        return src.crs, src.transform


def extract_raster_values(
    gdf: gpd.GeoDataFrame,
    raster_path: Path,
    col_name: str,
    nodata_val: float = None,
    reproject_if_needed: bool = True,
) -> pd.Series:
    """
    Extract raster values at settlement centroids.

    Handles:
      - CRS mismatch: reprojects points to raster CRS before sampling
      - NoData: replaces with NaN
      - Negative values: replaced with NaN (physical quantities are non-negative)

    Parameters
    ----------
    reproject_if_needed : if True, reprojects coordinates to match raster CRS
    """
    with rasterio.open(raster_path) as src:
        raster_crs = src.crs
        nd = src.nodata if nodata_val is None else nodata_val

        # Reproject points to raster CRS if needed
        if raster_crs and raster_crs.to_epsg() != 4326:
            pts = gpd.GeoDataFrame(geometry=gpd.points_from_xy(gdf["lon"], gdf["lat"]),
                                   crs="EPSG:4326").to_crs(raster_crs)
            coords = list(zip(pts.geometry.x, pts.geometry.y))
        else:
            coords = list(zip(gdf["lon"], gdf["lat"]))

        values = [v[0] for v in src.sample(coords)]

    series = pd.Series(values, index=gdf.index, name=col_name, dtype=float)
    if nd is not None:
        series = series.replace(nd, np.nan)
    series = series.where(series >= 0, np.nan)
    return series


def check_extraction(series: pd.Series, label: str, expected_min=None, expected_max=None):
    """Validate extracted values and warn if suspicious."""
    n_valid = series.notna().sum()
    n_total = len(series)
    print(f"  {label}: {n_valid:,}/{n_total:,} valid  |  "
          f"min={series.min():.3f}  mean={series.mean():.3f}  max={series.max():.3f}")
    if n_valid / n_total < 0.8:
        print(f"  ⚠ WARNING: {(1-n_valid/n_total)*100:.0f}% missing — check raster extent covers Benin")
    if series.nunique() <= 3:
        print(f"  ⚠ WARNING: Only {series.nunique()} unique values — possible nodata or CRS issue")
    if expected_min is not None and series.min() < expected_min:
        print(f"  ⚠ WARNING: min {series.min():.3f} below expected {expected_min}")
    if expected_max is not None and series.max() > expected_max:
        print(f"  ⚠ WARNING: max {series.max():.3f} above expected {expected_max}")
````

## 1. GHI — Global Horizontal Irradiance

**Column:** `GHI`  kWh/m²/day  
**Expected range for Benin:** 4.5 – 6.5 kWh/m²/day  
**Source:** NASA POWER, SolarGIS, PVGIS, or Global Solar Atlas (download as GeoTIFF)  
**Save to:** `data/raw/rasters/ghi_benin.tif`


````python
GHI_RASTER = FILES["ghi"]

if GHI_RASTER.exists():
    print("=== GHI RASTER DIAGNOSTICS ===")
    raster_diagnostics(GHI_RASTER, "GHI")
    print()

    gdf["GHI"] = extract_raster_values(gdf, GHI_RASTER, "GHI")
    check_extraction(gdf["GHI"], "GHI (kWh/m²/day)", expected_min=3.0, expected_max=8.0)

    # Fill any gaps with Benin latitude-band averages (rare edge settlements)
    gap_mask = gdf["GHI"].isna()
    if gap_mask.sum() > 0:
        print(f"  Filling {gap_mask.sum()} gaps with latitude-band average")
        lat_avg = gdf.groupby(gdf["lat"].round(1))["GHI"].transform("mean")
        gdf.loc[gap_mask, "GHI"] = lat_avg[gap_mask]

    print(f"\nGHI extraction complete ✓  ({gdf['GHI'].notna().sum():,} settlements)")

else:
    print("GHI raster NOT found.")
    print("  Download from: https://globalsolaratlas.info/download/benin")
    print("  Or NASA POWER: https://power.larc.nasa.gov/data-access-viewer/")
    print("  Save as: data/raw/rasters/ghi_benin.tif")
    print()
    print("  Using latitude-based proxy (Benin ranges 4.8–6.2 kWh/m²/day):")
    gdf["GHI"] = gdf["lat"].apply(lambda lat:
        6.2 if lat >= 11  else
        5.9 if lat >= 10  else
        5.6 if lat >= 9   else
        5.2 if lat >= 8   else
        4.9 if lat >= 7   else 4.8
    )
    print("  ⚠ Proxy active — replace with real raster before final analysis")

print(gdf["GHI"].describe().round(3))
````

## 2. Wind Speed

**Column:** `WindVel` (m/s), `WindCF` (capacity factor 0–1)  
**Expected range for Benin:** 2.4 – 5.4 m/s  
**Source:** Global Wind Atlas — download as GeoTIFF at 250m resolution  
**Save to:** `data/raw/rasters/windvelocity.tif`


````python
def wind_capacity_factor(ws: float, cut_in=3.0, rated=12.0, cut_out=25.0) -> float:
    """Simplified capacity factor from mean wind speed (cubic power law)."""
    if pd.isna(ws) or ws < cut_in or ws > cut_out:
        return 0.0
    return min((ws / rated) ** 3, 1.0)


WIND_RASTER = FILES["wind"]

if WIND_RASTER.exists():
    print("=== WIND RASTER DIAGNOSTICS ===")
    raster_diagnostics(WIND_RASTER, "WindVel")
    print()

    gdf["WindVel"] = extract_raster_values(gdf, WIND_RASTER, "WindVel")
    check_extraction(gdf["WindVel"], "Wind speed (m/s)", expected_min=1.0, expected_max=12.0)

    # Fill gaps
    gap_mask = gdf["WindVel"].isna()
    if gap_mask.sum() > 0:
        print(f"  Filling {gap_mask.sum()} gaps with spatial average")
        lat_avg = gdf.groupby(gdf["lat"].round(1))["WindVel"].transform("mean")
        gdf.loc[gap_mask, "WindVel"] = lat_avg[gap_mask]

    gdf["WindCF"] = gdf["WindVel"].apply(wind_capacity_factor)
    print(f"\nWind extraction complete ✓")
    print(f"  Settlements with WindVel ≥ 3 m/s (feasible): {(gdf['WindVel'] >= 3).sum():,} ({(gdf['WindVel'] >= 3).mean()*100:.1f}%)")

else:
    print("Wind raster NOT found.")
    print("  Download from: https://globalwindatlas.info  → Download area → Benin → Mean wind speed 50m")
    print("  Save as: data/raw/rasters/windvelocity.tif")
    print()
    print("  Using latitude-based proxy:")
    gdf["WindVel"] = gdf["lat"].apply(lambda lat:
        5.2 if lat >= 10 else
        4.5 if lat >= 9  else
        3.8 if lat >= 8  else
        3.2 if lat >= 7  else 2.8
    )
    gdf["WindCF"] = gdf["WindVel"].apply(wind_capacity_factor)
    print("  ⚠ Proxy active — replace with real raster before final analysis")

print(gdf[["WindVel","WindCF"]].describe().round(3))
````

## 3. Slope & Elevation (from DEM)

**Columns:** `Slope` (degrees), `Elevation` (m)  
**Expected range for Benin:** Slope 0–9°, Elevation 0–650 m  
**Source:** SRTM 30m via OpenTopography or EarthData  
**Workflow:**  
1. If `slope_benin.tif` exists → extract directly  
2. Else if `srtm_benin.tif` exists → compute slope, save, then extract  
3. Else → proxy  

**Save DEM to:** `data/raw/rasters/srtm_benin.tif`


````python
def compute_slope_from_dem(dem_path: Path, slope_path: Path) -> bool:
    """
    Compute slope (degrees) from DEM using finite differences.
    Masks nodata before gradient to avoid artefacts at raster edges.
    Cell size computed from actual pixel resolution at Benin mean latitude (9.3°N).
    """
    try:
        with rasterio.open(dem_path) as src:
            elev      = src.read(1).astype(np.float64)
            transform = src.transform
            nd        = src.nodata
            profile   = src.profile.copy()
            res_x     = abs(transform.a)  # degrees/pixel x
            res_y     = abs(transform.e)  # degrees/pixel y

        # Mask nodata (avoids giant artificial gradients at ocean/border edges)
        if nd is not None:
            elev = np.where(elev == nd, np.nan, elev)

        # Pixel size in metres — Benin mean latitude 9.3°N
        lat_rad  = np.radians(9.3)
        cell_x   = res_x * 111320 * np.cos(lat_rad)  # ~109,500 m/deg at 9°N
        cell_y   = res_y * 111320

        dy, dx    = np.gradient(elev, cell_y, cell_x)
        slope_deg = np.degrees(np.arctan(np.sqrt(dx**2 + dy**2)))
        slope_deg = np.clip(slope_deg, 0, 90)

        profile.update(dtype="float32", nodata=-9999)
        with rasterio.open(slope_path, "w", **profile) as dst:
            dst.write(slope_deg.astype("float32"), 1)

        valid = slope_deg[np.isfinite(slope_deg)]
        print(f"  Slope computed: min={valid.min():.2f}°  max={valid.max():.2f}°  mean={valid.mean():.2f}°")
        return True
    except Exception as e:
        print(f"  Slope computation failed: {e}")
        import traceback; traceback.print_exc()
        return False


SLOPE_RASTER = FILES["slope"]
DEM_RASTER   = FILES["dem"]

print("=== SLOPE / ELEVATION ===")

# Step 1: compute slope raster if needed
if not SLOPE_RASTER.exists() and DEM_RASTER.exists():
    print(f"DEM found ({DEM_RASTER.name}) — computing slope raster...")
    raster_diagnostics(DEM_RASTER, "DEM (elevation)")
    ok = compute_slope_from_dem(DEM_RASTER, SLOPE_RASTER)
    if not ok:
        SLOPE_RASTER = None  # force proxy
elif SLOPE_RASTER.exists():
    print(f"Pre-computed slope raster found: {SLOPE_RASTER.name}")
else:
    print("Neither slope nor DEM raster found — using proxy")
    SLOPE_RASTER = None

# Step 2: extract elevation from DEM
if DEM_RASTER.exists():
    print()
    print("Extracting elevation...")
    raster_diagnostics(DEM_RASTER, "Elevation (m)")
    gdf["Elevation"] = extract_raster_values(gdf, DEM_RASTER, "Elevation")
    check_extraction(gdf["Elevation"], "Elevation (m)", expected_min=0, expected_max=700)
else:
    print("No DEM — using latitude-based elevation proxy")
    def elev_proxy(lat, lon):
        if 9.0 <= lat <= 11.0 and 1.0 <= lon <= 2.5: return np.random.uniform(200, 500)
        elif lat >= 10: return np.random.uniform(150, 350)
        elif lat >= 7:  return np.random.uniform(50, 200)
        else:           return np.random.uniform(0, 80)
    gdf["Elevation"] = gdf.apply(lambda r: elev_proxy(r["lat"], r["lon"]), axis=1)

# Step 3: extract slope
if SLOPE_RASTER is not None and Path(str(SLOPE_RASTER)).exists():
    print()
    print("Extracting slope from raster...")
    raster_diagnostics(SLOPE_RASTER, "Slope (°)")
    gdf["Slope"] = extract_raster_values(gdf, SLOPE_RASTER, "Slope")
    check_extraction(gdf["Slope"], "Slope (°)", expected_min=0, expected_max=45)
else:
    print("Using slope proxy: flat assumption (2°) for most of Benin")
    gdf["Slope"] = 2.0
    print("  ⚠ Proxy active — provide srtm_benin.tif for real values")

print()
print("Elevation:", gdf["Elevation"].describe().round(1).to_dict())
print("Slope:    ", gdf["Slope"].describe().round(3).to_dict())
````

## 4. Hydro Potential (vector point layer)

**Columns:** `HydropowerDist` (km), `Hydropower` (kW), `HydroHead` (m), `HydroDischarge` (m³/s)  
**Your data schema:** `fid, OBJECTID, river_ID, elev, head, discharge_, power (W), PowerMW`  
**Save to:** `data/raw/hydro_potential_benin.shp`  (or `.geojson` / `.gpkg`)

For each settlement: find nearest hydro site, record distance and capacity.  
Feasibility gate in `mini_grid.py`: distance ≤ 5 km AND Hydropower ≥ 10 kW.


````python
def detect_power_kw(hydro: gpd.GeoDataFrame) -> pd.Series:
    """
    Auto-detect power column and return values in kW.
    Handles: power (Watts), PowerMW (megawatts), power_kw, capacity_kw.
    Detection logic: if max > 10,000 → treat as Watts, divide by 1000.
    """
    cols_lower = {c.lower(): c for c in hydro.columns}

    if "power_kw"    in cols_lower: return hydro[cols_lower["power_kw"]].fillna(0).astype(float)
    if "capacity_kw" in cols_lower: return hydro[cols_lower["capacity_kw"]].fillna(0).astype(float)

    if "power" in cols_lower:
        raw = hydro[cols_lower["power"]].fillna(0).astype(float)
        if raw.max() > 10000:
            print(f"  power column in Watts (max={raw.max():,.0f} W) → dividing by 1000 → kW")
            return raw / 1000.0
        return raw

    if "powermw" in cols_lower:
        raw = hydro[cols_lower["powermw"]].fillna(0).astype(float)
        print(f"  PowerMW column (max={raw.max():.4f} MW) → multiplying by 1000 → kW")
        return raw * 1000.0

    print("  ⚠ No power column found — Hydropower set to 0")
    return pd.Series(0.0, index=hydro.index)


def hydro_spatial_join(
    gdf: gpd.GeoDataFrame,
    hydro_path: Path,
    search_radius_km: float = 15.0,
) -> gpd.GeoDataFrame:
    """
    For each settlement, find the nearest hydro potential site within search_radius_km.
    Sites beyond this radius → HydropowerDist = actual distance, Hydropower = 0 (infeasible).

    The 5 km mini-hydro feasibility threshold is enforced in mini_grid.py, not here.
    We use 15 km search radius to capture near-miss sites for the sensitivity analysis.
    """
    hydro = gpd.read_file(hydro_path).to_crs("EPSG:4326")

    print(f"  {len(hydro):,} hydro sites loaded from {hydro_path.name}")
    print(f"  Columns: {list(hydro.columns)}")

    # Detect and convert power to kW
    hydro["power_kw"] = detect_power_kw(hydro)

    # Dataset diagnostics
    print(f"\n  Power range : {hydro['power_kw'].min():.1f} – {hydro['power_kw'].max():.1f} kW")
    if "head" in hydro.columns:
        print(f"  Head range  : {hydro['head'].min():.1f} – {hydro['head'].max():.1f} m")
    if "discharge_" in hydro.columns:
        print(f"  Discharge   : {hydro['discharge_'].min():.2f} – {hydro['discharge_'].max():.2f} m³/s")
    print(f"  Sites ≥ 10 kW  : {(hydro['power_kw'] >= 10).sum():,}")
    print(f"  Sites ≥ 100 kW : {(hydro['power_kw'] >= 100).sum():,}")
    print(f"  Sites ≥ 1 MW   : {(hydro['power_kw'] >= 1000).sum():,}")

    # Project to UTM 31N for metre-accurate distances
    gdf_m   = gdf.to_crs("EPSG:32631")
    hydro_m = hydro.to_crs("EPSG:32631")

    hydro_union = hydro_m.geometry.unary_union
    print(f"\n  Running nearest-site join for {len(gdf_m):,} settlements...")

    def nearest_hydro(point):
        near_geom = nearest_points(point, hydro_union)[1]
        dist_m    = point.distance(near_geom)
        idx       = hydro_m.geometry.distance(point).idxmin()
        site      = hydro_m.loc[idx]
        return {
            "HydropowerDist" : round(dist_m / 1000, 3),
            "Hydropower"     : float(site["power_kw"]),
            "HydropowerFID"  : int(idx),
            "HydroHead"      : float(site["head"])        if "head"       in site.index else np.nan,
            "HydroDischarge" : float(site["discharge_"])  if "discharge_" in site.index else np.nan,
        }

    results  = gdf_m.geometry.apply(lambda pt: pd.Series(nearest_hydro(pt)))
    gdf      = gdf.copy()
    for col in ["HydropowerDist","Hydropower","HydropowerFID","HydroHead","HydroDischarge"]:
        gdf[col] = results[col].values

    # Zero out sites beyond search radius (too far to be viable)
    beyond = gdf["HydropowerDist"] > search_radius_km
    gdf.loc[beyond, "Hydropower"] = 0.0
    print(f"  Within {search_radius_km} km: {(~beyond).sum():,}  |  Beyond: {beyond.sum():,}")
    return gdf


HYDRO_FILE = FILES["hydro"]

print("=== HYDRO POTENTIAL SPATIAL JOIN ===")
if HYDRO_FILE.exists():
    gdf = hydro_spatial_join(gdf, HYDRO_FILE, search_radius_km=15.0)

    # Resource statistics only — feasibility threshold applied in Notebook 02
    for thresh_km in [2, 5, 10, 15]:
        n = (gdf["HydropowerDist"] <= thresh_km).sum()
        print(f"  Settlements within {thresh_km:>2} km of a hydro site : {n:,} ({n/len(gdf)*100:.1f}%)")
    print()
    print(gdf[["HydropowerDist","Hydropower","HydroHead","HydroDischarge"]].describe().round(2))
else:
    print("Hydro file NOT found.")
    print("  Save as: data/raw/hydro_potential_benin.shp  (or .geojson / .gpkg)")
    print("  Using river-distance proxy...")
    dist_col = next((c for c in ["dist_lake_river_km","closest_distance_water","dist_river_km"]
                     if c in gdf.columns), None)
    if dist_col:
        gdf["HydropowerDist"] = gdf[dist_col].fillna(999)
        gdf["Hydropower"]     = gdf["HydropowerDist"].apply(
            lambda d: 150.0 if d <= 2 else 30.0 if d <= 5 else 0.0)
    else:
        gdf["HydropowerDist"] = 999.0
        gdf["Hydropower"]     = 0.0
    gdf["HydropowerFID"]  = -1
    gdf["HydroHead"]      = np.nan
    gdf["HydroDischarge"] = np.nan
    print("  ⚠ Proxy active")
````

## 5. Travel Time to Nearest City

**Primary source: Oxford MAP accessibility raster** — `accessibility_cities_benin.tif`  
Point-sampled at each settlement centroid, giving travel time in **minutes** to the
nearest city ≥50,000 population (Weiss et al. 2020 motorized travel time surface).

> *Weiss D.J. et al. (2020). Global maps of travel time to healthcare facilities.*
> *Nature Medicine 26, 1835–1838. DOI: 10.1038/s41591-020-1059-1*

**Validation: comparison against VIDA `dist_nearest_hub_km`**  
VIDA provides straight-line distance to nearest OSM town (`dist_nearest_hub_km`).
These two accessibility proxies measure different things:

| Measure | Column | Units | Captures |
|---|---|---|---|
| MAP raster (this cell) | `TravelHours` | hours | Road quality, terrain, speed limits |
| VIDA hub distance | `CityDistKm` | km | Euclidean/road distance only |

A scatter plot and Pearson correlation are computed to assess agreement.
Settlements where they diverge strongly flag poor road conditions or
settlements near a city but cut off by terrain/rivers.


````python
# ── Travel Time — MAP Accessibility Raster ───────────────────────────────────
# Source: Weiss et al. 2020, Oxford Malaria Atlas Project
# File:   data/raw/rasters/accessibility_cities_benin.tif
# Values: minutes of motorized travel time to nearest city >=50,000 pop
# OnSSET uses this exact dataset for its TravelHours input column

ACCESS_RASTER = FILES['access']

if ACCESS_RASTER.exists():
    print('=== ACCESSIBILITY RASTER DIAGNOSTICS ===')
    raster_diagnostics(ACCESS_RASTER, 'Travel time (min)')

    # Point-sample raster at settlement centroids
    gdf['travel_time_raster_min'] = extract_raster_values(
        gdf, ACCESS_RASTER, 'AccessibilityMin'
    )

    # Clip negative nodata artefacts and fill gaps with median
    gdf['travel_time_raster_min'] = gdf['travel_time_raster_min'].clip(lower=0)
    median_tt = gdf['travel_time_raster_min'].median()
    n_gaps    = gdf['travel_time_raster_min'].isna().sum()
    gdf['travel_time_raster_min'] = gdf['travel_time_raster_min'].fillna(median_tt)

    gdf['TravelHours'] = gdf['travel_time_raster_min'] / 60.0

    print(f'\n  Extracted from raster  ✓')
    print(f'  Gaps filled (median {median_tt:.0f} min): {n_gaps} settlements')
    print(f'  Range : {gdf.travel_time_raster_min.min():.0f} – '
          f'{gdf.travel_time_raster_min.max():.0f} min')
    print(f'  Median: {gdf.travel_time_raster_min.median():.0f} min  '
          f'({gdf.TravelHours.median():.2f} hr)')

    # ── VIDA comparison ────────────────────────────────────────────────────
    # VIDA also provides travel_time_min (pre-computed from the same MAP surface)
    # and dist_nearest_hub_km (Euclidean distance to nearest OSM town)
    print('\n=== COMPARISON: MAP RASTER vs VIDA COLUMNS ===')

    import matplotlib.pyplot as plt
    import matplotlib.gridspec as gridspec
    from scipy import stats

    fig = plt.figure(figsize=(14, 5))
    gs  = gridspec.GridSpec(1, 2, figure=fig, wspace=0.35)

    # --- Panel A: Raster vs VIDA travel_time_min ---
    ax1 = fig.add_subplot(gs[0])
    if 'travel_time_min' in gdf.columns:
        vida_tt = pd.to_numeric(gdf['travel_time_min'], errors='coerce').fillna(median_tt)
        r1, p1  = stats.pearsonr(
            gdf['travel_time_raster_min'].dropna(),
            vida_tt[gdf['travel_time_raster_min'].notna()]
        )
        ax1.scatter(gdf['travel_time_raster_min'], vida_tt,
                    alpha=0.15, s=4, color='steelblue', rasterized=True)
        mn = min(gdf['travel_time_raster_min'].max(), vida_tt.max())
        ax1.plot([0, mn], [0, mn], 'r--', lw=1, label='1:1 line')
        ax1.set_xlabel('MAP Raster — travel time (min)', fontsize=10)
        ax1.set_ylabel('VIDA travel_time_min (min)', fontsize=10)
        ax1.set_title(f'Raster vs VIDA travel time\nPearson r = {r1:.3f}  (p < 0.001)',
                      fontsize=10)
        ax1.legend(fontsize=8)
        print(f'  Panel A — Raster vs VIDA travel_time_min : r = {r1:.3f}')
    else:
        ax1.text(0.5, 0.5, 'travel_time_min\nnot in VIDA data',
                 ha='center', va='center', transform=ax1.transAxes)
        print('  Panel A — VIDA travel_time_min column not found')

    # --- Panel B: TravelHours vs dist_nearest_hub_km ---
    ax2 = fig.add_subplot(gs[1])
    if 'dist_nearest_hub_km' in gdf.columns:
        hub_km  = pd.to_numeric(gdf['dist_nearest_hub_km'], errors='coerce').dropna()
        tt_hrs  = gdf.loc[hub_km.index, 'TravelHours']
        r2, p2  = stats.pearsonr(tt_hrs, hub_km)

        ax2.scatter(hub_km, tt_hrs,
                    alpha=0.15, s=4, color='darkorange', rasterized=True)

        # Mark outliers: high travel time but low distance (bad roads / terrain)
        outlier_mask = (tt_hrs > tt_hrs.quantile(0.9)) & (hub_km < hub_km.quantile(0.3))
        if outlier_mask.sum() > 0:
            ax2.scatter(hub_km[outlier_mask], tt_hrs[outlier_mask],
                        s=12, color='red', alpha=0.5, label=f'Outliers (n={outlier_mask.sum()})')
            ax2.legend(fontsize=8)

        ax2.set_xlabel('VIDA dist_nearest_hub_km (km)', fontsize=10)
        ax2.set_ylabel('MAP Raster — TravelHours (hr)', fontsize=10)
        ax2.set_title(f'City distance vs travel time\nPearson r = {r2:.3f}  (p < 0.001)',
                      fontsize=10)
        print(f'  Panel B — TravelHours vs dist_nearest_hub_km   : r = {r2:.3f}')

        # Regression slope: implied average speed
        slope, intercept, *_ = stats.linregress(hub_km, tt_hrs)
        implied_speed = 1 / slope  # km per hour
        print(f'  Implied avg travel speed: {implied_speed:.1f} km/h '
              f'(road quality indicator)')
    else:
        ax2.text(0.5, 0.5, 'dist_nearest_hub_km\nnot in VIDA data',
                 ha='center', va='center', transform=ax2.transAxes)

    fig.suptitle('Travel Time Accessibility — MAP Raster vs VIDA Columns\n'
                 'Benin Settlements (n=17,205)', fontsize=11, fontweight='bold')
    plt.savefig(BASE_DIR / 'data' / 'processed' / 'travel_time_comparison.png',
                dpi=150, bbox_inches='tight')
    plt.show()
    print('  Plot saved → data/processed/travel_time_comparison.png')

else:
    # ── Fallback: use VIDA pre-computed column ──────────────────────────────
    warnings.warn(
        'Accessibility raster not found — falling back to VIDA travel_time_min.\n'
        'Place file at: data/raw/rasters/accessibility_cities_benin.tif'
    )
    if 'travel_time_min' in gdf.columns:
        gdf['travel_time_raster_min'] = pd.to_numeric(
            gdf['travel_time_min'], errors='coerce'
        ).fillna(60.0)
    else:
        gdf['travel_time_raster_min'] = 60.0
        warnings.warn('travel_time_min also missing — using 60 min constant fallback')

    gdf['TravelHours'] = gdf['travel_time_raster_min'] / 60.0
    print('  TravelHours from VIDA fallback (raster preferred)')
    print(f'  Median: {gdf.TravelHours.median():.2f} hr')
````

## 7. Grid Distance

**Source: VIDA/DRE Atlas settlement data** — World Bank energydata.info grid lines  
*(VIDA Technical Documentation, Section 2: 'Access to electricity', p.12)*

| VIDA column | Official definition (VIDA Annex) | Used as |
|---|---|---|
| `distance_to_existing_transmission_lines` | Distance to closest existing MV/HV line | **`GridDistKm`** — primary LCOE input |
| `distance_to_planned_transmission_lines` | Distance to closest planned MV/HV line | `DistPlannedLine` — grid priority bonus |
| `dist_to_substations` | Distance to nearest substation | `DistSubstation` — secondary reference |
| `dist_nearest_hub_km` | Distance to nearest city/town (OSM) | `CityDistKm` — **market signal only, NOT grid** |
| `dist_lake_river_km` | Distance to closest river or lake (HydroSHEDS) | `dist_river_km` — mini-hydro feasibility |
| `dist_main_road_km` | Distance to closest main road (OSM) | `dist_road_km` — hybrid MG feasibility |

> ⚠ **VIDA note:** *"publicly available information on the existing and planned
> transmission grid can be incomplete and/or outdated"* — treat `GridDistKm` as
> indicative; validate against SBEE/ABERME national grid maps where possible.


````python
# Grid / transmission distance — read directly from VIDA settlement attributes
#
# PRIMARY:  distance_to_existing_transmission_lines  → GridDistKm (used in LCOE)
# BONUS:    distance_to_planned_transmission_lines   → DistPlannedLine (grid priority score)
# SUBST:    dist_to_substations                      → DistSubstation (secondary cost check)
# CITY:     nearest_hub_km / dist_to_regional_hubs   → CityDistKm (demand signal, NOT grid)

# Primary grid distance for LCOE — existing transmission line
gdf['GridDistKm'] = pd.to_numeric(
    gdf['distance_to_existing_transmission_lines'], errors='coerce'
).fillna(gdf['distance_to_existing_transmission_lines'].median() if
         pd.to_numeric(gdf['distance_to_existing_transmission_lines'],
                       errors='coerce').notna().any() else 25.0)

# Planned line distance (grid priority score bonus: settlements within 10km get ×1.25)
gdf['DistPlannedLine'] = pd.to_numeric(
    gdf['distance_to_planned_transmission_lines'], errors='coerce'
).fillna(999)

# Substation distance (alternative grid cost reference)
gdf['DistSubstation'] = pd.to_numeric(
    gdf['dist_to_substations'], errors='coerce'
).fillna(999)

# City/urban hub distance — market demand signal, NOT used for grid LCOE
# VIDA column is 'dist_nearest_hub_km' (distance to nearest city/town — OSM)
gdf['CityDistKm'] = pd.to_numeric(
    gdf['dist_nearest_hub_km'], errors='coerce'
).fillna(50.0)

# River distance for mini-hydro feasibility
gdf['dist_river_km'] = pd.to_numeric(
    gdf['dist_lake_river_km'], errors='coerce'
).fillna(999)

# Road distance for hybrid mini-grid feasibility
gdf['dist_road_km'] = pd.to_numeric(
    gdf['dist_main_road_km'], errors='coerce'
).fillna(999)

# Grid proximity buckets (based on existing line distance)
gdf['GridProximity'] = pd.cut(
    gdf['GridDistKm'],
    bins=[0, 5, 15, 25, 9999],
    labels=['<5 km', '5–15 km', '15–25 km', '>25 km']
)

print('=== GRID & INFRASTRUCTURE DISTANCES ===')
for col, label in [
    ('GridDistKm',      'Existing line (km) [→ LCOE]  '),
    ('DistPlannedLine', 'Planned line (km)  [→ bonus]  '),
    ('DistSubstation',  'Substation (km)               '),
    ('CityDistKm',      'Nearest city (km)  [demand]   '),
    ('dist_river_km',   'River/lake (km)    [hydro MG] '),
    ('dist_road_km',    'Main road (km)     [hybrid MG]'),
]:
    s = gdf[col]
    valid = s[s < 999]
    print(f'  {label}: median {valid.median():.1f}  range {valid.min():.1f}–{valid.max():.1f}  '
          f'(n={valid.notna().sum():,})')

print(f'\nGrid proximity (existing line):')
print(gdf['GridProximity'].value_counts().sort_index().to_string())
print(f'\nWithin 25 km grid threshold: '
      f'{(gdf.GridDistKm <= 25).sum():,} / {len(gdf):,} settlements')
````

## 6. Derived: Diesel Transport Cost

**Method: OnSSET transport logistics model** *(gep_onsset.py, line 760)*

OnSSET computes the delivered diesel price as the pump price plus a transport adder
that reflects how much fuel the delivery truck burns per litre it delivers:

```
DieselPrice (USD/L) = p_d + 2 × p_d × C_truck × TravelHours / V_truck
```

| Parameter | Symbol | Value | Source |
|---|---|---|---|
| Base pump price | `p_d` | 0.85 USD/L | SBEE/SONEEB regulated price, Benin 2023 |
| Truck consumption | `C_truck` | 33.7 L/hr | OnSSET `gep_runner.py` (large tanker) |
| Truck tank volume | `V_truck` | 15,000 L | OnSSET `gep_runner.py` (large tanker) |
| Travel time | `TravelHours` | per settlement | MAP raster (cell 15) — minutes / 60 |
| Round-trip factor | `2×` | — | OnSSET convention (outbound + return) |

The `2×` accounts for the round trip. The effective transport coefficient is:
`2 × 0.85 × 33.7 / 15,000 ≈ 0.00382 USD/L per travel-hour`
— much smaller than a flat ×0.04 assumption, and physically grounded in truck logistics.


````python
# Diesel delivered price — OnSSET transport logistics model
# Source: gep_onsset.py line 760 + gep_runner.py lines 255-256
#
# Formula:  p_delivered = p_d + 2 * p_d * C_truck * TravelHours / V_truck
#   p_d      = base pump price (USD/litre)
#   C_truck  = truck fuel consumption (litres/hour)  — 33.7 L/hr large tanker
#   V_truck  = truck tank volume (litres)            — 15,000 L large tanker
#   2×       = round-trip (outbound + return journey)
#   LHV_DIESEL = 9.9445 kWh/litre (used downstream in LCOE, stored here for reference)

DIESEL_BASE_USD_L  = 0.85       # USD/litre — SBEE/SONEEB regulated, Benin 2023
TRUCK_CONSUMPTION  = 33.7       # L/hr  — large tanker (OnSSET gep_runner.py)
TRUCK_VOLUME       = 15_000     # litres — large tanker (OnSSET gep_runner.py)
LHV_DIESEL         = 9.9445     # kWh/litre (OnSSET constant, used in LCOE notebook)

# Effective transport coefficient (USD/L per travel-hour)
K_transport = 2 * DIESEL_BASE_USD_L * TRUCK_CONSUMPTION / TRUCK_VOLUME

gdf['DieselPrice'] = DIESEL_BASE_USD_L + K_transport * gdf['TravelHours']

print('=== DIESEL DELIVERED PRICE — OnSSET logistics model ===')
print(f'  Base pump price  : ${DIESEL_BASE_USD_L:.3f} /L')
print(f'  Transport coeff  : ${K_transport:.5f} /L per travel-hour')
print(f'  (vs. flat 0.04 assumption — OnSSET value is ~10× smaller & physically grounded)')
print()
print(gdf['DieselPrice'].describe().round(4).to_string())
print(f"\n  Min delivered: ${gdf['DieselPrice'].min():.4f}/L  "
      f"(TravelHours={gdf['TravelHours'].min():.2f})")
print(f"  Max delivered: ${gdf['DieselPrice'].max():.4f}/L  "
      f"(TravelHours={gdf['TravelHours'].max():.2f})")
print(f"  Premium at median travel ({gdf['TravelHours'].median():.2f} hr): "
      f"${K_transport * gdf['TravelHours'].median():.4f}/L "
      f"({K_transport * gdf['TravelHours'].median() / DIESEL_BASE_USD_L * 100:.2f}% above base)")
````

## 8. Night-Time Lights — VIIRS 2020

**File:** `data/raw/rasters/Benin_VIIRS_nighttime_lights_2020.tif`  
**Source:** VIIRS DNB monthly composites (NASA/NOAA)  
**Unit:** nanoWatts/cm²/sr — proxy for existing electrification level

Used for:
- `NightLights` column in OnSSET (electrification status proxy)
- Cross-validating VIDA `has_nightlight` flag
- Identifying already-electrified settlements to exclude from analysis

**Extraction method:** max value in a 3×3 pixel window around each settlement centroid  
(single-pixel sampling misses settlements that fall between pixel centres at 500m resolution)

````python
import rasterio
from rasterio.windows import Window

VIIRS_CANDIDATES = [
    BASE_DIR / 'data' / 'raw' / 'rasters' / 'Benin_VIIRS_nighttime_lights_2020.tif',
    BASE_DIR / 'data' / 'raw' / 'rasters' / 'Benin_VIIRS_nighttime_lights_2020.tif',
    BASE_DIR / 'data' / 'raw' / 'rasters' / 'viirs_benin_2020.tif',
    BASE_DIR / 'data' / 'raw' / 'rasters' / 'nighttime_lights.tif',
]
VIIRS_RASTER = next((p for p in VIIRS_CANDIDATES if p.exists()), None)


def extract_viirs_max3x3(gdf: gpd.GeoDataFrame, raster_path: Path) -> pd.Series:
    """
    Extract VIIRS night-time light value at each settlement using a
    3x3 pixel neighbourhood maximum.

    Why 3x3 and not single-pixel?
    VIIRS DNB is 500m resolution. A settlement point can fall anywhere
    within a 500m cell, or straddle a cell boundary. Taking the max of
    the 3x3 window (1.5km neighbourhood) avoids systematically under-
    counting electrification for settlements near pixel edges.

    Returns
    -------
    pd.Series of float — NTL radiance in nW/cm²/sr, NaN where no data
    """
    results = []
    with rasterio.open(raster_path) as src:
        nodata = src.nodata
        for _, row in gdf.iterrows():
            try:
                # Get pixel coordinates for this point
                py, px = src.index(row['lon'], row['lat'])
                # 3x3 window centred on pixel
                win = Window(
                    col_off = max(0, px - 1),
                    row_off = max(0, py - 1),
                    width   = min(3, src.width  - max(0, px - 1)),
                    height  = min(3, src.height - max(0, py - 1))
                )
                patch = src.read(1, window=win).astype(float)
                if nodata is not None:
                    patch = patch[patch != nodata]
                patch = patch[patch >= 0]
                results.append(float(patch.max()) if len(patch) > 0 else np.nan)
            except Exception:
                results.append(np.nan)
    return pd.Series(results, index=gdf.index, name='NightLights')


if VIIRS_RASTER is not None:
    print(f'VIIRS raster found: {VIIRS_RASTER.name}')

    # Diagnostics
    with rasterio.open(VIIRS_RASTER) as src:
        band  = src.read(1).astype(float)
        nd    = src.nodata
        valid = band[(band != nd) & (band >= 0)] if nd is not None else band[band >= 0]
        print(f'  CRS         : {src.crs}')
        print(f'  Resolution  : {abs(src.transform.a)*111320:.0f} m/pixel (approx)')
        print(f'  NoData      : {nd}')
        print(f'  Value range : {valid.min():.2f} – {valid.max():.2f} nW/cm²/sr')
        print(f'  Lit pixels  : {(valid > 0).sum():,} of {len(valid):,} ({(valid>0).mean()*100:.1f}%)')

    print('Extracting NTL values (3x3 max window)...')
    gdf['NightLights'] = extract_viirs_max3x3(gdf, VIIRS_RASTER)

    n_lit    = (gdf['NightLights'] > 0).sum()
    n_bright = (gdf['NightLights'] > 5).sum()
    print(f'\nNight-lights extracted: {gdf["NightLights"].notna().sum():,} valid values')
    print(f'  Settlements with NTL > 0  (any light)    : {n_lit:,} ({n_lit/len(gdf)*100:.1f}%)')
    print(f'  Settlements with NTL > 5  (likely elec.) : {n_bright:,} ({n_bright/len(gdf)*100:.1f}%)')

    # Cross-validate against VIDA has_nightlight if present
    if 'has_nightlight' in gdf.columns:
        vida_lit  = gdf['has_nightlight'].astype(bool)
        viirs_lit = gdf['NightLights'] > 0
        agree = (vida_lit == viirs_lit).mean()
        print(f'\nCross-validation vs VIDA has_nightlight:')
        print(f'  Agreement: {agree*100:.1f}%')
        print(f'  VIDA lit: {vida_lit.sum():,} | VIIRS lit: {viirs_lit.sum():,}')

else:
    print('VIIRS raster NOT found')
    print(f'  Expected: data/raw/rasters/Benin_VIIRS_nighttime_lights_2020.tif')
    print(f'  Your QGIS layer is: Benin_VIIRS_nighttime_lights_2020')
    print(f'  → Right-click layer in QGIS → Export → Save As → GeoTIFF')
    gdf['NightLights'] = np.nan

print(f'\nNight-lights stats:')
print(gdf['NightLights'].describe().round(3))
````

## 9. Land Cover — MODIS MCD12Q1 2022

**File:** `data/raw/rasters/MODIS_MCD12Q1v061_Benin_2022.tif`  
**Source:** MODIS Terra+Aqua Land Cover Type, 500m, annual (NASA LP DAAC)  
**Classification:** IGBP Land Cover scheme (Type 1), integer class codes

| Code | Class | Relevance |
|------|-------|-----------|
| 1–5  | Forest types | Low productive use demand |
| 10   | Grasslands | Agropastoral — moderate demand |
| 12   | Croplands | Agricultural productive use — higher demand |
| 14   | Cropland/Natural mosaic | Common in Benin south |
| 13   | Urban & built-up | Commercial/service sector — highest demand |
| 16   | Barren | Low demand |

**Used for:**
- Productive use demand adjustment (see `ProductiveUseFactor` below)
- Cross-validating VIDA building density
- OnSSET `LandCover` column input

---

### Productive Use Factor — Justification & Literature Evidence

The `ProductiveUseFactor` is a demand multiplier applied on top of VIDA's satellite-derived
residential demand to account for non-household electricity uses associated with each land
cover type. It is applied as: `adjusted_demand = VIDA_demand × ProductiveUseFactor`.

| Factor | Value | Basis |
|--------|-------|-------|
| Cropland (IGBP 12, 14) | **×1.25** | Kirchen et al. (2023); Farthing et al. (2023); GEP 2.0 (World Bank, 2021) |
| Urban / built-up (IGBP 13) | **×1.40** | GEP-OnSSET urban tier logic; Benin sectoral demand split |
| All other classes | **×1.00** | Baseline residential demand only |

#### Cropland ×1.25 — Key References

1. **Kirchen et al. (2023)** — *"Energy planning in Sub-Saharan African countries needs to
   explicitly consider productive uses of electricity"*, Scientific Reports, 13, 13053.
   [https://doi.org/10.1038/s41598-023-40021-y](https://doi.org/10.1038/s41598-023-40021-y)  
   > Shows that productive activities — irrigation, agro-processing (milling, husking,
   > cassava grinding), and small-scale manufacturing — **increase household electricity
   > demand by approximately 50%** across SSA. The ×1.25 factor used here is a conservative
   > lower bound of this range.

2. **Farthing et al. (2023)** — *"Quantifying agricultural productive use of energy load in
   Sub-Saharan Africa and its impact on microgrid configurations and costs"*,
   Applied Energy, 340, 121015.
   [https://doi.org/10.1016/j.apenergy.2023.121015](https://doi.org/10.1016/j.apenergy.2023.121015)  
   > Develops a 10×10 km resolution geospatial methodology using **cropland extent as a
   > primary driver** of agricultural PUE demand (irrigation, milling, shelling, oil pressing).
   > Directly supports land-cover as a spatial proxy for productive use load.

3. **Stevanović et al. (2024)** — *"The potential of electric agriculture and mobility for
   the least-cost rural electrification strategy in Sub-Saharan Africa"*,
   Energy for Sustainable Development, 83, 101583.
   [https://doi.org/10.1016/j.esd.2024.101583](https://doi.org/10.1016/j.esd.2024.101583)  
   > Finds that **rural areas with significant cropland particularly benefit** from
   > electrified agricultural technologies. Including cropland-linked demand shifts the
   > least-cost technology mix toward mini-grids (60% vs. 13% without PUE in Chad).

4. **Domegni & Azouma (2022)** — *"Productive uses of energy: A solution for promoting
   energy justice in rural areas in West Africa"*, Renewable & Sustainable Energy Reviews,
   159, 112155. **Includes 640 surveyed households in Benin.**
   [https://doi.org/10.1016/j.rser.2022.112155](https://doi.org/10.1016/j.rser.2022.112155)  
   > Surveys across Benin, Senegal, and Togo show **53.7% of rural economic activity is
   > agricultural/pastoralist**. Off-grid electrification accounting only for household
   > demand is financially unsustainable without integrating productive use potential.

5. **World Bank / ESMAP — GEP 2.0 (2021)**
   [https://blogs.worldbank.org/en/energy/the-global-electrification-platform-3-0--more-data-for-more-accu](https://blogs.worldbank.org/en/energy/the-global-electrification-platform-3-0--more-data-for-more-accu)  
   > GEP 2.0 extended OnSSET to include commercial and agricultural demand sectors. In
   > Madagascar, adding these non-residential demands increased estimated access
   > requirements from 1.6 GW to **2.6 GW** — a 63% uplift — confirming that ignoring
   > land-use-linked productive demand systematically undersizes electrification plans.

#### Urban / Built-up ×1.40 — Key References

6. **GEP-OnSSET source code — `SET_COMMERCIAL_DEMAND` / urban tier parameters**
   [https://github.com/global-electrification-platform/gep-onsset](https://github.com/global-electrification-platform/gep-onsset)  
   > GEP-OnSSET explicitly separates urban and rural target demand tiers and maintains
   > distinct demand columns for commercial, health, and educational activities
   > (`SET_COMMERCIAL_DEMAND`, `SET_EDU_DEMAND`, `SET_HEALTH_DEMAND`), reflecting that
   > urban built-up areas carry structurally higher energy loads than rural settlements.

7. **Benin Energy Situation (Energypedia)**
   [https://energypedia.info/wiki/Benin_Energy_Situation](https://energypedia.info/wiki/Benin_Energy_Situation)  
   > In Benin, **households account for 48%** of total electricity consumption, the
   > institutional sector 32%, and the industrial/commercial sector 24% — confirming
   > that urban built-up settlements carry significant non-residential demand that
   > residential-only estimates would miss entirely.

#### Note on conservatism
The ×1.25 cropland factor is **deliberately conservative**: Kirchen et al. (2023) report
~50% uplift when full productive use is realised. The ×1.25 reflects partial adoption of
PUE consistent with Benin's current low mechanisation rate and nascent agro-processing
sector. Sensitivity testing with ×1.40 and ×1.50 is recommended for the scenario analysis
in Notebook 07.

---

### Relationship to VIDA's Agricultural Data Approach

The VIDA/DRE Atlas uses **MapSPAM 2017** (IFPRI Spatial Production Allocation Model)
for regional agricultural indicators — extracted at **administrative subdivision level 2**
(not settlement level) due to MapSPAM's 10 km pixel resolution.
*(VIDA Technical Doc, p.13: "Regional agriculture indicators")*

The `ProductiveUseFactor` derived from MODIS land cover in this notebook is a
**settlement-level enhancement** of that regional signal — providing spatial granularity
that MapSPAM cannot offer at 500 m resolution. The two approaches are complementary:

| Data | Resolution | Used for |
|---|---|---|
| MapSPAM (VIDA/DRE Atlas) | ~10 km / admin level 2 | Crop type, value, yield context |
| MODIS MCD12Q1 (this notebook) | 500 m / settlement level | `ProductiveUseFactor` demand uplift |

**Cross-validation:** Where MapSPAM shows high crop value per hectare in a district AND
MODIS shows cropland (code 12/14) at the settlement level, the ×1.25 factor is
well-supported by both sources.


````python
MODIS_CANDIDATES = [
    BASE_DIR / 'data' / 'raw' / 'rasters' / 'MODIS_MCD12Q1v061_Benin_2022.tif',
    BASE_DIR / 'data' / 'raw' / 'rasters' / 'modis_landcover_benin_2022.tif',
    BASE_DIR / 'data' / 'raw' / 'rasters' / 'landcover_benin.tif',
    BASE_DIR / 'data' / 'raw' / 'rasters' / 'land_cover_benin.tif',
]
MODIS_RASTER = next((p for p in MODIS_CANDIDATES if p.exists()), None)

# IGBP class labels for MODIS MCD12Q1 Type 1
IGBP_LABELS = {
    1:  'Evergreen Needleleaf Forest',
    2:  'Evergreen Broadleaf Forest',
    3:  'Deciduous Needleleaf Forest',
    4:  'Deciduous Broadleaf Forest',
    5:  'Mixed Forest',
    6:  'Closed Shrublands',
    7:  'Open Shrublands',
    8:  'Woody Savannas',
    9:  'Savannas',
    10: 'Grasslands',
    11: 'Permanent Wetlands',
    12: 'Croplands',
    13: 'Urban & Built-up',
    14: 'Cropland/Natural Vegetation Mosaic',
    15: 'Snow & Ice',
    16: 'Barren or Sparsely Vegetated',
    17: 'Water Bodies',
    255:'Unclassified',
}

# Classes associated with agricultural productive use
PRODUCTIVE_USE_CLASSES = {12, 14}   # Croplands + Cropland mosaic
URBAN_CLASSES          = {13}


def productive_use_factor(land_class: int) -> float:
    """
    Demand uplift factor based on land cover.
    Cropland settlements tend to have higher productive use loads
    (milling, irrigation, cold storage) than purely pastoral ones.
    """
    if land_class in PRODUCTIVE_USE_CLASSES: return 1.25   # +25% demand
    if land_class in URBAN_CLASSES:          return 1.40   # +40% demand
    return 1.0                                             # no adjustment


if MODIS_RASTER is not None:
    print(f'MODIS land cover raster found: {MODIS_RASTER.name}')

    # Diagnostics
    with rasterio.open(MODIS_RASTER) as src:
        band   = src.read(1)
        nd     = src.nodata
        valid  = band[band != nd] if nd is not None else band.flatten()
        unique = np.unique(valid)
        print(f'  CRS         : {src.crs}')
        print(f'  NoData      : {nd}')
        print(f'  Unique codes: {sorted(unique.tolist())}')

    # Extract using nearest-pixel (mode of 3x3 would be ideal but slower)
    gdf['LandCover'] = extract_raster_values(gdf, MODIS_RASTER, 'LandCover',
                                              nodata_val=255).fillna(255).astype(int)
    gdf['LandCoverLabel'] = gdf['LandCover'].map(IGBP_LABELS).fillna('Unknown')
    gdf['ProductiveUseFactor'] = gdf['LandCover'].apply(productive_use_factor)

    print(f'\nLand cover distribution across settlements:')
    dist = gdf['LandCoverLabel'].value_counts()
    for label, count in dist.items():
        pct = count / len(gdf) * 100
        bar = '█' * int(pct / 2)
        print(f'  {label:<40} {count:5,} ({pct:4.1f}%) {bar}')

    pu = (gdf['ProductiveUseFactor'] > 1.0).sum()
    print(f'\nSettlements with productive use uplift (cropland): {pu:,} ({pu/len(gdf)*100:.1f}%)')

    # Cross-validate against VIDA building density
    if 'Medium_and_large_buildings_pc' in gdf.columns:
        print('\nCropland vs building density cross-check:')
        cropland_mask = gdf['LandCover'].isin(PRODUCTIVE_USE_CLASSES)
        print(f'  Mean large-building% in cropland   : {gdf.loc[cropland_mask,"Medium_and_large_buildings_pc"].mean():.3f}')
        print(f'  Mean large-building% in non-cropland: {gdf.loc[~cropland_mask,"Medium_and_large_buildings_pc"].mean():.3f}')

else:
    print('MODIS raster NOT found')
    print(f'  Expected: data/raw/rasters/MODIS_MCD12Q1v061_Benin_2022.tif')
    print(f'  Your QGIS layer is: MODIS_MCD12Q1v061_Benin_2022')
    print(f'  → Right-click layer in QGIS → Export → Save As → GeoTIFF')
    gdf['LandCover']          = 255   # Unclassified
    gdf['LandCoverLabel']     = 'Unclassified'
    gdf['ProductiveUseFactor'] = 1.0

print(f'\nLand cover extraction complete')
````

**No feasibility flags are computed here** — technology eligibility and system sizing
are handled downstream in Notebook 02 (LCOE) and the technology selector.

|---|---|---|
| Solar irradiance | `GHI` | Solar MG & SHS LCOE, capacity factor |
| Wind speed | `WindVel`, `WindCF` | Wind MG LCOE |
| Slope / Elevation | `Slope`, `Elevation` | Grid extension cost, site suitability |

| Diesel price | `DieselPrice` | Hybrid MG LCOE |

| Land cover | `LandCover`, `LandCoverLabel`, `ProductiveUseFactor` | Demand adjustment |


````python
import matplotlib.pyplot as plt

# ── Resource layer coverage summary ──────────────────────────────────────────
# No feasibility flags here — technology selection handled in Notebook 02

print('=== RENEWABLE ENERGY RESOURCE LAYERS ===')
print(f"  {'Layer':<22} {'Unit':<14} {'Min':>8}  {'Mean':>8}  {'Max':>8}  {'Valid':>12}")
print('  ' + '─' * 78)

resource_layers = [
    ('GHI',            'kWh/m²/day'),
    ('WindVel',        'm/s'),
    ('WindCF',         'fraction'),
    ('Slope',          'degrees'),
    ('Elevation',      'm'),
    ('Hydropower',     'kW'),
    ('HydropowerDist', 'km'),
    ('HydroHead',      'm'),
    ('HydroDischarge', 'm³/s'),
    ('TravelHours',    'hours'),
    ('DieselPrice',    'USD/litre'),
    ('GridDistKm',     'km'),
    ('DistPlannedLine','km'),
    ('dist_river_km',  'km'),
    ('dist_road_km',   'km'),
    ('NightLights',    'nW/cm²/sr'),
]

for col, unit in resource_layers:
    if col in gdf.columns:
        s = pd.to_numeric(gdf[col], errors='coerce').dropna()
        valid_str = f'{s.notna().sum():,}/{len(gdf):,}'
        print(f"  {col:<22} {unit:<14} {s.min():>8.2f}  {s.mean():>8.2f}  "
              f"{s.max():>8.2f}  {valid_str:>12}")
    else:
        print(f"  {col:<22} {'':14} {'NOT IN DATAFRAME':>42}")

# Categorical columns
print('\n=== CATEGORICAL / BOOLEAN LAYERS ===')
for col in ['LandCoverLabel', 'ProductiveUseFactor', 'GridProximity', 'has_nightlight']:
    if col in gdf.columns:
        print(f"\n  {col}:")
        print(gdf[col].value_counts().to_string())

# ── Distribution plots (raw resource values only) ─────────────────────────
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
fig.suptitle('Renewable Energy Resource Distributions — Benin Settlements',
             fontsize=13, fontweight='bold')

plot_layers = [
    ('GHI',           'GHI (kWh/m²/day)',     'skyblue',   axes[0,0]),
    ('WindVel',       'Wind speed (m/s)',      'steelblue', axes[0,1]),
    ('Slope',         'Slope (degrees)',       'peru',      axes[0,2]),
    ('Elevation',     'Elevation (m)',         'olive',     axes[0,3]),
    ('Hydropower',    'Hydro capacity (kW)',   'teal',      axes[1,0]),
    ('HydropowerDist','Hydro distance (km)',   'cadetblue', axes[1,1]),
    ('TravelHours',   'Travel time (hours)',   'salmon',    axes[1,2]),
    ('GridDistKm',    'Grid distance (km)',    'slategray', axes[1,3]),
]
for col, xlabel, color, ax in plot_layers:
    if col in gdf.columns:
        data = pd.to_numeric(gdf[col], errors='coerce').dropna()
        ax.hist(data, bins=50, color=color, edgecolor='white', alpha=0.85)
        ax.axvline(data.mean(), color='red', linestyle='--', lw=1.2,
                   label=f'mean={data.mean():.2f}')
        ax.set_xlabel(xlabel)
        ax.set_ylabel('Settlements')
        ax.legend(fontsize=8)
        ax.set_title(col)
    else:
        ax.set_visible(False)

plt.tight_layout()
out_dir = BASE_DIR / 'data' / 'outputs' / 'maps'
out_dir.mkdir(parents=True, exist_ok=True)
plt.savefig(out_dir / 'resource_layer_distributions.png', dpi=150)
plt.show()
print('\nDistribution plot saved ✓')
````
