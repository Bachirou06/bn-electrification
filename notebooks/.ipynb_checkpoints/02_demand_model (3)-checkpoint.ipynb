# 02 — Demand Model
**Benin Least-Cost Electrification Analysis**

## Demand Strategy: VIDA Satellite Data as Primary Source

Unlike standard electrification tools that assign a national average demand tier
to all settlements, this analysis uses **VIDA's satellite-derived building footprint
data** to give every settlement its own unique demand estimate.

### VIDA's Demand Formula (Reverse-Engineered)
VIDA computes `energy_demand` (kWh/day) using building sizes detected from satellite imagery:

```
energy_demand (kWh/day) = small_buildings  × 0.2170
                        + medium_buildings × 0.6500
                        + large_buildings  × 1.1000

Annual (kWh/year)       = energy_demand × 365
```
Validation: OLS regression, **r = 1.000, MAE = 0.000** — perfect fit confirmed.

### Productive Use Uplift (from Notebook 00 enriched data)
Final demand is scaled by `ProductiveUseFactor` derived from MODIS land cover:

| Land cover | Factor | Basis |
|---|---|---|
| Cropland (IGBP 12, 14) | ×1.25 | Kirchen et al. (2023); GEP 2.0 World Bank |
| Urban / built-up (IGBP 13) | ×1.40 | GEP-OnSSET urban tier; Benin appliance mix |
| All other | ×1.00 | Baseline residential |

### Social Infrastructure Bonus
Settlements with a **health facility** or **education facility** receive a +1 MTF tier
upgrade in the fallback model, reflecting higher anchor-load demand.

### Demand Growth
Base demand is grown at **4%/year** over the 15-year planning horizon.

### Fallback
Settlements where VIDA `energy_demand = 0` (< 1% of total) fall back to the MTF
tier estimate.


````python
import sys
sys.path.append('..')

import geopandas as gpd
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.demand.demand_estimator import add_demand_columns
from src.demand.mtf_tiers import tier_summary
from src.config import CONFIG, GENERAL, DEMAND

sns.set_theme(style='whitegrid')
print('Libraries loaded ✓')
````

## 1. Load Settlement Data

Loads the **GIS-enriched file** from Notebook 00 (`settlements_gis_enriched.geojson`).
This provides `ProductiveUseFactor`, `LandCoverLabel`, `has_health_facility`, and `has_education_facility`
alongside the core VIDA demand columns.


````python
from pathlib import Path

BASE_DIR      = Path('..').resolve()
PROCESSED_DIR = BASE_DIR / 'data' / 'processed'
RAW_DIR       = BASE_DIR / 'data' / 'raw'

# Load GIS-enriched output of Notebook 00 (has ProductiveUseFactor + social cols)
enriched_candidates = sorted(
    list(PROCESSED_DIR.glob('settlements_gis_enriched*.geojson')),
    reverse=True
)
if enriched_candidates:
    SETTLEMENTS_PATH = enriched_candidates[0]
    print(f'Loading enriched file: {SETTLEMENTS_PATH.name}')
else:
    SETTLEMENTS_PATH = RAW_DIR / 'Benin_settlement_properties.geojson'
    print('⚠ Enriched file not found — loading raw VIDA (ProductiveUseFactor unavailable)')
    print('  Run Notebook 00 first for full demand model.')

gdf = gpd.read_file(SETTLEMENTS_PATH)
print(f'Loaded {len(gdf):,} settlements  |  {len(gdf.columns)} columns')

# Confirm key columns present
key_cols = ['energy_demand', 'ProductiveUseFactor',
            'has_health_facility', 'has_education_facility',
            'mean_rwi', 'Medium_and_large_buildings_pc']
print('\nKey column check:')
for col in key_cols:
    status = '✓' if col in gdf.columns else '✗ MISSING'
    print(f'  {col:<35}: {status}')
````

````python
# ── Productive Use Factor & Social Infrastructure preview ─────────────────────

print('=== PRODUCTIVE USE FACTOR (from MODIS land cover) ===')
if 'ProductiveUseFactor' in gdf.columns:
    puf = gdf['ProductiveUseFactor'].value_counts().sort_index()
    for factor, count in puf.items():
        pct = count / len(gdf) * 100
        label = {1.0: 'Baseline (residential)',
                 1.25: 'Cropland uplift',
                 1.40: 'Urban/built-up uplift'}.get(factor, '')
        print(f'  ×{factor:.2f}  {label:<28}: {count:6,} settlements ({pct:.1f}%)')
    lc_col = 'LandCoverLabel' if 'LandCoverLabel' in gdf.columns else None
    if lc_col:
        print('\nLand cover breakdown:')
        print(gdf[lc_col].value_counts().head(8).to_string())
else:
    print('  ⚠ ProductiveUseFactor not found — will default to ×1.00')
    gdf['ProductiveUseFactor'] = 1.0

print('\n=== SOCIAL INFRASTRUCTURE ===')
for col, label in [('has_health_facility',    'Health facility'),
                    ('has_education_facility', 'Education facility')]:
    if col in gdf.columns:
        n   = gdf[col].sum()
        pct = n / len(gdf) * 100
        print(f'  {label:<25}: {n:,} settlements ({pct:.1f}%)')
    else:
        print(f'  {label:<25}: column not found')
````

## 2. Verify VIDA Demand Columns Are Present

````python
vida_cols = ['energy_demand', 'avg_connection_energy_demand',
             'small_buildings', 'medium_buildings', 'large_buildings',
             'num_connections', 'mean_rwi', 'Medium_and_large_buildings_pc']

print('VIDA demand columns check:')
for col in vida_cols:
    if col in gdf.columns:
        n_valid = gdf[col].notna().sum()
        pct     = n_valid / len(gdf) * 100
        print(f'  ✓ {col:40s} — {n_valid:,} / {len(gdf):,} valid ({pct:.1f}%)')
    else:
        print(f'  ✗ {col:40s} — MISSING')

# ── Outlier detection & cap ───────────────────────────────────────────────
# Must run BEFORE formula validation so the scatter plot is not dominated
# by extreme values. Cap at 99th percentile.
ed   = gdf['energy_demand']
p99  = ed.quantile(0.99)
p999 = ed.quantile(0.999)
n_outliers = (ed > p99).sum()

print(f'\n=== OUTLIER CHECK ===')
print(f'  Median          : {ed.median():.1f} kWh/day')
print(f'  Mean            : {ed.mean():.1f} kWh/day'
      f'{"  ⚠ mean >> median" if ed.mean() > 3*ed.median() else ""}')
print(f'  99th percentile : {p99:.1f} kWh/day')
print(f'  Max             : {ed.max():.1f} kWh/day')
print(f'  Above p99       : {n_outliers:,} settlements ({n_outliers/len(gdf)*100:.2f}%)')

if ed.max() > 10_000:
    print(f'\n  ⚠ Capping energy_demand at p99 ({p99:.0f} kWh/day) — '
          f'max {ed.max():.0f} is implausible for a single settlement')
    print(f'  Top 5 outliers before cap:')
    top5 = gdf.nlargest(5, 'energy_demand')[['village_name','population',
                                               'num_connections','energy_demand']]
    print(top5.to_string(index=False))

    # Cap all building-count columns proportionally
    cap_ratio = gdf['energy_demand'].clip(upper=p99) / gdf['energy_demand'].replace(0, np.nan)
    for col in ['small_buildings','medium_buildings','large_buildings',
                'energy_demand','avg_connection_energy_demand']:
        if col in gdf.columns:
            gdf[col] = (gdf[col] * cap_ratio).fillna(gdf[col]).clip(upper=gdf[col].max())
    gdf['energy_demand'] = gdf['energy_demand'].clip(upper=p99)
    print(f'\n  Capped ✓  new max = {gdf["energy_demand"].max():.1f} kWh/day')

n_zero = (gdf['energy_demand'] == 0).sum()
print(f'\n  Zero-demand settlements: {n_zero:,} → '
      f'{"MTF fallback will apply" if n_zero > 0 else "no fallback needed"}')
````

## 3. Validate VIDA's Building Formula
Confirm the reverse-engineered formula: `energy_demand = small×0.217 + medium×0.650 + large×1.100`

````python
required = ['energy_demand', 'small_buildings', 'medium_buildings', 'large_buildings']
if all(c in gdf.columns for c in required):
    predicted = (gdf['small_buildings']  * 0.2170
               + gdf['medium_buildings'] * 0.6500
               + gdf['large_buildings']  * 1.1000)
    r   = gdf['energy_demand'].corr(predicted)
    mae = (gdf['energy_demand'] - predicted).abs().mean()

    print('VIDA Building Formula Validation (after outlier cap)')
    print(f'  energy_demand = small×0.217 + medium×0.650 + large×1.100')
    print(f'  Correlation r = {r:.4f}  (expected ~1.000)')
    print(f'  MAE           = {mae:.4f} kWh/day  (expected ~0.000)')
    print(f'  N settlements = {len(gdf):,}')

    fig, ax = plt.subplots(figsize=(7, 5))
    ax.scatter(gdf['energy_demand'], predicted,
               alpha=0.15, s=4, color='#1E88E5', rasterized=True)
    lim = max(gdf['energy_demand'].max(), predicted.max()) * 1.05
    ax.plot([0, lim], [0, lim], 'r--', lw=1.5,
            label=f'Perfect fit (r={r:.4f})')
    ax.set_xlabel('VIDA energy_demand (kWh/day)')
    ax.set_ylabel('Building formula prediction (kWh/day)')
    ax.set_title('Validation: VIDA Building Formula\n(outliers capped at p99)')
    ax.legend(fontsize=9)
    plt.tight_layout()
    plt.savefig('../data/outputs/maps/vida_formula_validation.png', dpi=150)
    plt.show()
else:
    print('Building columns not available — skipping formula validation')
````

## 4. Run Demand Model (VIDA Primary + MTF Cross-Validation)

````python
gdf = add_demand_columns(gdf)
print('Demand estimation complete ✓')

vida_count = (gdf['demand_source'] == 'vida').sum()
mtf_count  = (gdf['demand_source'] == 'mtf_fallback').sum()
print(f'\nDemand source breakdown:')
print(f'  VIDA satellite data  : {vida_count:,} ({vida_count/len(gdf)*100:.1f}%)')
print(f'  MTF fallback         : {mtf_count:,} ({mtf_count/len(gdf)*100:.1f}%)')

# ── Apply ProductiveUseFactor uplift ─────────────────────────────────────
if 'ProductiveUseFactor' in gdf.columns:
    gdf['demand_year0_kwh'] = gdf['demand_year0_kwh'] * gdf['ProductiveUseFactor']
    gdf['demand_yearT_kwh'] = gdf['demand_yearT_kwh'] * gdf['ProductiveUseFactor']
    if 'vida_demand_year0_kwh' in gdf.columns:
        gdf['vida_demand_year0_kwh'] = gdf['vida_demand_year0_kwh'] * gdf['ProductiveUseFactor']
    uplifted = (gdf['ProductiveUseFactor'] > 1.0).sum()
    print(f'\nProductiveUseFactor applied:')
    print(f'  Uplifted settlements : {uplifted:,} ({uplifted/len(gdf)*100:.1f}%)')
    print(f'  ×1.25 cropland       : {(gdf.ProductiveUseFactor==1.25).sum():,}')
    print(f'  ×1.40 urban          : {(gdf.ProductiveUseFactor==1.40).sum():,}')
else:
    print('\n⚠ ProductiveUseFactor not applied (column missing)')

print(f'\nDemand at Year 0 (kWh/year) — after productive use uplift:')
print(gdf['demand_year0_kwh'].describe().round(0))
````

## 3b. How RWI Feeds into the Model

The **Relative Wealth Index (RWI)** plays a specific, bounded role alongside VIDA's building data:

| Role | Module | What it controls |
|---|---|---|
| **MTF tier assignment** | `mtf_tiers.py` | RWI + building mix + services → Tier 1–4 |
| **SHS kit sizing** | `shs.py` | Tier 1–2 → $150 Tier-2 kit; Tier 3+ → $350 Tier-3 kit |
| **LCOE denominator** | `shs.py` | VIDA demand_timeseries / num_households (not kit rated kWh) |
| **Priority equity score** | `technology_selector.py` | Poorer settlements ranked higher |

**Key distinction**: RWI determines *which SHS kit size to recommend* (a product/affordability signal),
while VIDA's building data determines *how much electricity the settlement needs* (the demand signal).
These are separate inputs to the LCOE formula.

### Tier scoring (all inputs from VIDA):
```
score = 0
if mean_rwi >= 0.5:                        score += 2   # high wealth
elif mean_rwi >= 0.0:                      score += 1   # medium wealth
if Medium_and_large_buildings_pc >= 0.30:  score += 2   # many large buildings
elif Medium_and_large_buildings_pc >= 0.10: score += 1
if has_health_facility or has_education:   score += 1   # anchor loads present

Tier: 0-1 pts → T1 | 2 pts → T2 | 3 pts → T3 | 4-5 pts → T4
SHS kit: T1-T2 → $150 Tier-2 | T3-T4 → $350 Tier-3
```

````python
# Show RWI distribution and its effect on tier + SHS kit assignment
# mtf_tier now exists (created by add_demand_columns above)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
fig.suptitle('RWI Role: Tier Assignment & SHS Kit Sizing', fontsize=12, fontweight='bold')

# 1. RWI distribution with tier thresholds
rwi_vals = gdf['mean_rwi'].dropna()
axes[0].hist(rwi_vals, bins=60, color='#5C6BC0', edgecolor='white')
axes[0].axvline(0.5, color='red', linestyle='--', label='High threshold (0.5)')
axes[0].axvline(0.0, color='orange', linestyle='--', label='Medium threshold (0.0)')
axes[0].set_title('RWI Distribution')
axes[0].set_xlabel('Relative Wealth Index')
axes[0].set_ylabel('Settlements')
axes[0].legend(fontsize=8)
n_high = (rwi_vals >= 0.5).sum()
n_med  = ((rwi_vals >= 0.0) & (rwi_vals < 0.5)).sum()
n_low  = (rwi_vals < 0.0).sum()
axes[0].text(0.98, 0.97,
             f'High (>=0.5): {n_high:,}\nMed (0–0.5): {n_med:,}\nLow (<0.0): {n_low:,}',
             transform=axes[0].transAxes, va='top', ha='right', fontsize=8,
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

# 2. MTF tier distribution
tier_counts = gdf['mtf_tier'].value_counts().sort_index()
tier_colors = {1: '#FFF9C4', 2: '#FFE082', 3: '#FFB300', 4: '#E65100'}
axes[1].bar(tier_counts.index, tier_counts.values,
            color=[tier_colors[t] for t in tier_counts.index], edgecolor='gray')
axes[1].set_title('MTF Tier Distribution\n(RWI + building mix + services)')
axes[1].set_xlabel('MTF Tier')
axes[1].set_ylabel('Settlements')
for t, n in tier_counts.items():
    axes[1].text(t, n + 30, f'{n:,}\n({n/len(gdf)*100:.0f}%)', ha='center', fontsize=8)

# 3. SHS kit mapping
shs_mapping = gdf['mtf_tier'].apply(lambda t: '$350 Tier-3 kit' if t >= 3 else '$150 Tier-2 kit')
shs_counts  = shs_mapping.value_counts()
axes[2].bar(shs_counts.index, shs_counts.values, color=['#4CAF50', '#FF9800'])
axes[2].set_title('SHS Kit Assignment\n(from RWI-informed MTF tier)')
axes[2].set_ylabel('Settlements')
for i, (k, v) in enumerate(shs_counts.items()):
    axes[2].text(i, v + 30, f'{v:,}\n({v/len(gdf)*100:.0f}%)', ha='center', fontsize=9)

plt.tight_layout()
plt.savefig('../data/outputs/maps/rwi_tier_shs_mapping.png', dpi=150)
plt.show()

print('RWI → Tier → SHS kit summary:')
for tier in sorted(tier_counts.index):
    n   = tier_counts[tier]
    kit = '$350 Tier-3' if tier >= 3 else '$150 Tier-2'
    print(f'  Tier {tier}: {n:,} settlements → SHS {kit} kit')
````

## 5. Cross-Validate: VIDA Demand vs MTF Tier Estimates

````python
# Compare VIDA demand with what MTF tiers would predict
vida_sub = gdf[gdf['demand_source'] == 'vida'].copy()

# What MTF would predict at t=0
from src.demand.mtf_tiers import get_tier_demand
vida_sub['mtf_demand_kwh'] = vida_sub.apply(
    lambda r: (r['num_households'] *
               DEMAND['electrification_rate_initial'] *
               get_tier_demand(int(r['mtf_tier']))),
    axis=1
)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('VIDA Demand vs MTF Tier Estimates — Cross-Validation', fontsize=13, fontweight='bold')

# Scatter: VIDA vs MTF
axes[0].scatter(vida_sub['mtf_demand_kwh'], vida_sub['demand_year0_kwh'],
                alpha=0.15, s=5, color='#FF9800')
lim = max(vida_sub[['mtf_demand_kwh','demand_year0_kwh']].quantile(0.95).max(), 1)
axes[0].plot([0, lim], [0, lim], 'r--', alpha=0.7, label='1:1 line')
axes[0].set_xlabel('MTF Tier Demand (kWh/year)')
axes[0].set_ylabel('VIDA Demand (kWh/year)')
axes[0].set_title('VIDA vs MTF per Settlement')
axes[0].legend()

# Box: VIDA demand by MTF tier
tier_labels = {1: 'T1', 2: 'T2', 3: 'T3', 4: 'T4'}
tier_groups = [vida_sub[vida_sub['mtf_tier'] == t]['demand_year0_kwh']
               for t in [1,2,3,4]]
axes[1].boxplot(tier_groups, labels=['Tier 1','Tier 2','Tier 3','Tier 4'],
                showfliers=False, patch_artist=True,
                boxprops=dict(facecolor='#90CAF9'))
axes[1].set_ylabel('VIDA Demand (kWh/year)')
axes[1].set_title('VIDA Demand Distribution by MTF Tier')
axes[1].set_yscale('log')

plt.tight_layout()
plt.savefig('../data/outputs/maps/vida_vs_mtf_validation.png', dpi=150)
plt.show()

corr = vida_sub['demand_year0_kwh'].corr(vida_sub['mtf_demand_kwh'])
print(f'Correlation VIDA vs MTF demand: r = {corr:.3f}')
print('(Correlation > 0.6 confirms MTF tiers are consistent with VIDA data)')
````

## 6. Demand Summary Statistics

````python
print('=== DEMAND MODEL SUMMARY ===')
print(f'Planning horizon  : {GENERAL["planning_horizon_years"]} years')
print(f'Base year         : {GENERAL["base_year"]}')
print(f'Demand growth     : {DEMAND["demand_growth_rate"]*100:.0f}%/year')
print(f'Primary source    : VIDA energy_demand (kWh/day × 365)')
print()

total_y0  = gdf['demand_year0_kwh'].sum() / 1e6
total_yT  = gdf['demand_yearT_kwh'].sum() / 1e6
print(f'Total national demand  Year 0  : {total_y0:.1f} GWh/year')
print(f'Total national demand  Year 15 : {total_yT:.1f} GWh/year')
print(f'Growth factor                  : ×{total_yT/total_y0:.2f}')
print()

summary = gdf.groupby('mtf_tier').agg(
    N_settlements       = ('demand_source', 'count'),
    VIDA_primary        = ('demand_source', lambda x: (x=='vida').sum()),
    Avg_VIDA_demand_kWh = ('vida_demand_year0_kwh', 'mean'),
    Avg_final_demand_kWh= ('demand_year0_kwh', 'mean'),
    Total_demand_GWh    = ('demand_year0_kwh', lambda x: x.sum()/1e6),
).round(1)
print('Demand by MTF Tier (cross-validation):')
print(summary.to_string())
````
